```python
# Part 1: Environment, Dataset Loading and Exploratory Data Analysis

import importlib.util, subprocess, sys

REQUIRED = {  # import name -> pip package name
    "numpy": "numpy",
    "pandas": "pandas",
    "scipy": "scipy",
    "statsmodels": "statsmodels",
    "matplotlib": "matplotlib",
    "seaborn": "seaborn",
    "wooldridge": "wooldridge",
}

missing = [pip_name for mod, pip_name in REQUIRED.items() if importlib.util.find_spec(mod) is None]

if missing:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *missing])

# ============================================================================
# Imports
try:
    import numpy as np
    import pandas as pd
    import scipy
    from scipy import stats
    import statsmodels
    import statsmodels.api as sm
    import matplotlib
    import matplotlib.pyplot as plt
    import seaborn as sns
    import wooldridge as woo
    from IPython.display import display
    from pathlib import Path
    from importlib.metadata import version, PackageNotFoundError
except ImportError as err:
    raise ImportError(
        f"A required package could not be imported: {err}. "
        "Re-run Cell 2, then restart the runtime and run from Cell 3."
    ) from err

# Display and plotting settings
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 130)
pd.set_option("display.float_format", lambda v: f"{v:,.3f}")
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({
    "figure.dpi": 110, "savefig.dpi": 300, "axes.titlesize": 12, "axes.titleweight": "bold",
    "axes.labelsize": 10, "legend.fontsize": 9, "figure.autolayout": False,
})
BLUE, RED, GREY = "#2F5D8A", "#B04A3F", "#6B6B6B"

# Folder for saving publication-quality PNGs of every figure
FIG_DIR = Path("figures_part1"); FIG_DIR.mkdir(exist_ok=True)
def save_fig(fig, name):
    fig.savefig(FIG_DIR / f"{name}.png", dpi=300, bbox_inches="tight")

print("Package versions")
for pkg in ["numpy", "pandas", "scipy", "statsmodels", "matplotlib", "seaborn", "wooldridge"]:
    try:
        print(f"  {pkg:12s} {version(pkg)}")
    except PackageNotFoundError:
        print(f"  {pkg:12s} (version not reported)")

# ============================================================================
# Load datasets with the package's documented function, woo.data(name)
def load_wooldridge_dataset(name):
    try:
        data = woo.data(name)
    except Exception as err:
        print(f"[ERROR] Could not load dataset '{name}' with woo.data('{name}').")
        print(f"  Exception type : {type(err).__name__}")
        print(f"  Message        : {err}")
        print("  Next steps:")
        print("    1. Upgrade the package:   !pip install --upgrade wooldridge   then restart the runtime.")
        print("    2. List available datasets: woo.data()   (check the spelling of the dataset name).")
        print(f"    3. Package location: {getattr(woo, '__file__', 'unknown')}")
        print("  No substitute dataset has been used. Fix the problem above before continuing.")
        raise
    if not isinstance(data, pd.DataFrame):
        raise TypeError(f"woo.data('{name}') returned {type(data).__name__}, expected a pandas DataFrame.")
    return data

wage1_raw = load_wooldridge_dataset("wage1")
phillips_raw = load_wooldridge_dataset("phillips")

# Fingerprints let us prove at the end that the raw copies were never altered
raw_fingerprint = {
    "wage1_raw": int(pd.util.hash_pandas_object(wage1_raw, index=True).sum()),
    "phillips_raw": int(pd.util.hash_pandas_object(phillips_raw, index=True).sum()),
}

# Independent working copies
wage1 = wage1_raw.copy(deep=True)
phillips = phillips_raw.copy(deep=True)

# Record of initial sample sizes (kept for the whole project)
initial_n = {"wage1": wage1_raw.shape[0], "phillips": phillips_raw.shape[0]}
print("Initial sample sizes:", initial_n)

# ============================================================================
# Basic verification of both datasets
for label, df in [("wage1", wage1), ("phillips", phillips)]:
    print("=" * 80)
    print(f"{label}: type = {type(df).__name__} | is pandas DataFrame: {isinstance(df, pd.DataFrame)}")
    print(f"Dimensions: {df.shape[0]} rows x {df.shape[1]} columns")
    print("Column names:", list(df.columns))
    print("\nData types:")
    display(df.dtypes.rename("dtype").to_frame().T)
    print("First five rows:")
    display(df.head())

# ============================================================================
# Print the documentation bundled with the package
for name in ["wage1", "phillips"]:
    print("#" * 80)
    woo.data(name, description=True)   # prints the documentation (returns None)
    print()

# Documented variable labels, transcribed from the output above for reuse in tables/plots
WAGE1_DOC = {
    "wage": "average hourly earnings", "educ": "years of education",
    "exper": "years potential experience", "tenure": "years with current employer",
    "nonwhite": "=1 if nonwhite", "female": "=1 if female", "married": "=1 if married",
    "numdep": "number of dependents", "smsa": "=1 if live in SMSA",
    "northcen": "=1 if live in north central U.S", "south": "=1 if live in southern region",
    "west": "=1 if live in western region", "construc": "=1 if work in construc. indus.",
    "ndurman": "=1 if in nondur. manuf. indus.", "trcommpu": "=1 if in trans, commun, pub ut",
    "trade": "=1 if in wholesale or retail", "services": "=1 if in services indus.",
    "profserv": "=1 if in prof. serv. indus.", "profocc": "=1 if in profess. occupation",
    "clerocc": "=1 if in clerical occupation", "servocc": "=1 if in service occupation",
    "lwage": "log(wage)", "expersq": "exper^2", "tenursq": "tenure^2",
}
PHILLIPS_DOC = {
    "year": "1948 through 2003", "unem": "civilian unemployment rate, %",
    "inf": "percentage change in CPI", "inf_1": "inf[_n-1]", "unem_1": "unem[_n-1]",
    "cinf": "inf - inf_1", "cunem": "unem - unem_1",
}
# Safety check: the documented variables should match the loaded columns
print("wage1 columns without a documented label    :", [c for c in wage1.columns if c not in WAGE1_DOC])
print("phillips columns without a documented label :", [c for c in phillips.columns if c not in PHILLIPS_DOC])

# ============================================================================
# Audit function (read-only: nothing is removed, imputed or changed)
def section(title):
    print("\n" + "=" * 80 + f"\n{title}\n" + "=" * 80)

def audit_dataset(df, name):
    res = {}
    section(f"{name}: size")
    res["n_obs"], res["n_vars"] = df.shape
    print(f"Observations (rows): {res['n_obs']}   |   Variables (columns): {res['n_vars']}")

    section(f"{name}: column names and data types")
    display(df.dtypes.rename("dtype").to_frame())

    section(f"{name}: first five and last five observations")
    display(df.head()); display(df.tail())

    section(f"{name}: missing values")
    miss = pd.DataFrame({"missing_count": df.isna().sum(), "missing_pct": df.isna().mean() * 100})
    display(miss)
    res["missing_cells"] = int(miss["missing_count"].sum())
    res["missing_by_col"] = miss.loc[miss["missing_count"] > 0, "missing_count"].to_dict()
    print(f"Total missing cells: {res['missing_cells']}")

    section(f"{name}: duplicate rows (identified, NOT deleted)")
    res["n_duplicate_rows"] = int(df.duplicated().sum())
    print(f"Rows that exactly repeat an earlier row: {res['n_duplicate_rows']}")
    if res["n_duplicate_rows"] > 0:
        display(df[df.duplicated(keep=False)].sort_values(list(df.columns)).head(20))
        print("Note: the data contain no respondent ID. Identical rows can be genuinely different "
              "people with the same recorded characteristics, so they are only flagged for inspection.")

    section(f"{name}: descriptive statistics (all numeric variables)")
    display(df.describe().T)

    section(f"{name}: infinite values and non-numeric content")
    numeric = df.select_dtypes(include=[np.number])
    inf_counts = np.isinf(numeric).sum()
    res["n_inf"] = int(inf_counts.sum())
    print(f"Infinite values found: {res['n_inf']}")
    non_numeric_cols = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c])]
    res["non_numeric_cols"] = non_numeric_cols
    print(f"Columns with a non-numeric dtype: {non_numeric_cols if non_numeric_cols else 'none'}")

    res["numeric_cols"] = list(numeric.columns)
    return res

def find_dummies(df):
    # A column is treated as a 0/1 indicator if its non-missing values are a subset of {0, 1}
    return [c for c in df.columns if set(df[c].dropna().unique()) <= {0, 1}]

audit_results = {}

# ============================================================================
# Run the audit on wage1
audit_results["wage1"] = audit_dataset(wage1, "wage1")

# ============================================================================
# Expected variables, frequency counts, suspicious-value flags (wage1)
EXPECTED = ["wage", "educ", "exper", "tenure", "female", "married", "nonwhite"]

section("wage1: presence of expected variables")
presence = pd.DataFrame({
    "present": [c in wage1.columns for c in EXPECTED],
    "documented_meaning": [WAGE1_DOC.get(c, "n/a") for c in EXPECTED],
}, index=EXPECTED)
display(presence)
absent = [c for c in EXPECTED if c not in wage1.columns]
if absent:
    print(f"[WARNING] Expected variables ABSENT from the loaded data: {absent}. "
          "Analyses that need them will be skipped or must be revised before Part 2.")
section("wage1: frequency counts for 0/1 indicator variables")
wage1_dummies = find_dummies(wage1)
print("Detected indicator variables:", wage1_dummies)
freq_rows = []
for c in wage1_dummies:
    vc = wage1[c].value_counts().reindex([0, 1], fill_value=0)
    freq_rows.append({"variable": c, "documented": WAGE1_DOC.get(c, ""), "count_0": vc[0], "count_1": vc[1],
                      "share_1_%": 100 * vc[1] / vc.sum()})
display(pd.DataFrame(freq_rows).set_index("variable"))

print("\nFrequency counts for the discrete count variable 'numdep' (number of dependents):")
display(wage1["numdep"].value_counts().sort_index().rename("count").to_frame().T)

section("wage1: plausibility flags (for INSPECTION only; no rows are removed)")
w = wage1
flags = {
    "wage <= 0": w["wage"] <= 0,
    "educ < 0 or educ > 20": (w["educ"] < 0) | (w["educ"] > 20),
    "exper < 0 or exper > 60": (w["exper"] < 0) | (w["exper"] > 60),
    "tenure < 0 or tenure > 50": (w["tenure"] < 0) | (w["tenure"] > 50),
    "tenure > exper (tenure exceeds potential experience)": w["tenure"] > w["exper"],
    "indicator value outside {0,1}": ~w[wage1_dummies].isin([0, 1]).all(axis=1),
    "lwage differs from log(wage) (tolerance 1e-6)": (w["lwage"] - np.log(w["wage"])).abs() > 1e-6,
    "expersq differs from exper^2": (w["expersq"] - w["exper"] ** 2).abs() > 1e-6,
    "tenursq differs from tenure^2": (w["tenursq"] - w["tenure"] ** 2).abs() > 1e-6,
}
flag_table = pd.DataFrame({"rows_flagged": {k: int(v.sum()) for k, v in flags.items()}})
display(flag_table)
print("Thresholds such as educ > 20 or exper > 60 are heuristic screening rules chosen for this audit, "
      "not rules from the dataset documentation.")
for k, mask in flags.items():
    if mask.sum() > 0:
        print(f"\nFlagged rows for: {k} (showing up to 10 of {int(mask.sum())})")
        display(w.loc[mask, ["wage", "educ", "exper", "tenure", "female", "married", "nonwhite"]].head(10))
audit_results["wage1"]["flag_counts"] = {k: int(v.sum()) for k, v in flags.items()}

# ============================================================================
# Run the audit on phillips
audit_results["phillips"] = audit_dataset(phillips, "phillips")

# ============================================================================
# Documented meanings, time-index information, plausibility flags (phillips)
section("phillips: documented variable meanings")
display(pd.DataFrame({"documented_meaning": pd.Series(PHILLIPS_DOC)}))
print("Source stated in the documentation: Economic Report of the President, 2004, Tables B-42 and B-64.")
print("Units/measurement beyond the labels above (e.g. which CPI series, annual average vs December) "
      "are NOT stated in the package documentation, so they are not assumed here.")

section("phillips: frequency counts for categorical/indicator variables")
ph_dummies = find_dummies(phillips)
print("Detected 0/1 indicator variables:", ph_dummies if ph_dummies else
      "none (all variables are continuous rates, a calendar year, or lagged/differenced copies)")

section("phillips: time-index information")
yr = phillips["year"]
doc_range = (1948, 2003)   # stated in the documentation label for 'year'
yr_whole = bool((yr.dropna() % 1 == 0).all())
yr_sorted = bool(yr.is_monotonic_increasing)
yr_unique = bool(yr.is_unique)
steps = sorted(yr.diff().dropna().unique().tolist())
yr_consecutive = (steps == [1])
print(f"Observed range of 'year'        : {yr.min():.0f} to {yr.max():.0f}  (documented: {doc_range[0]} through {doc_range[1]})")
print(f"Matches documented range        : {(yr.min(), yr.max()) == doc_range}")
print(f"Whole-number years              : {yr_whole}")
print(f"Sorted in increasing order      : {yr_sorted}")
print(f"No repeated years               : {yr_unique}")
print(f"Gaps between successive years   : {steps}   (consecutive annual steps only if this is [1.0])")
print(f"Number of observations          : {len(phillips)}")
phillips_time_ok = yr_whole and yr_sorted and yr_unique and yr_consecutive
print("\nConclusion: the 'year' variable is a documented time variable and the observations form a "
      "gap-free, ordered annual sequence." if phillips_time_ok else
      "\n[WARNING] The 'year' variable does not form a clean annual sequence. Verify ordering before Part 5.")

section("phillips: consistency of lagged/differenced variables with their documented definitions")
def max_abs_gap(a, b):
    return float((a - b).abs().max(skipna=True))
phillips_checks = pd.Series({
    "inf_1  vs inf shifted by one year":   max_abs_gap(phillips["inf_1"], phillips["inf"].shift(1)),
    "unem_1 vs unem shifted by one year":  max_abs_gap(phillips["unem_1"], phillips["unem"].shift(1)),
    "cinf   vs inf - inf_1":               max_abs_gap(phillips["cinf"], phillips["inf"] - phillips["inf_1"]),
    "cunem  vs unem - unem_1":             max_abs_gap(phillips["cunem"], phillips["unem"] - phillips["unem_1"]),
}, name="max absolute difference")
display(phillips_checks.to_frame())
phillips_defs_ok = bool((phillips_checks < 1e-5).all())   # tolerance allows for ~7-digit float rounding in the stored data
print("All derived variables reproduce their documented definitions." if phillips_defs_ok else
      "[WARNING] Some derived variables do not match their documented definitions.")

section("phillips: where are the missing values?")
miss_rows = phillips[phillips.isna().any(axis=1)]
print(f"Rows containing at least one missing value: {len(miss_rows)}")
if len(miss_rows):
    display(miss_rows)
    print("Lagged and differenced variables cannot exist for the first observation of a series, "
          "so missing values there arise from construction. They are NOT imputed.")

section("phillips: plausibility flags (inspection only)")
ph_flags = {
    "unem < 0 or unem > 100": (phillips["unem"] < 0) | (phillips["unem"] > 100),
    "|inf| > 20 (heuristic screen)": phillips["inf"].abs() > 20,
}
display(pd.DataFrame({"rows_flagged": {k: int(v.sum()) for k, v in ph_flags.items()}}))
audit_results["phillips"]["flag_counts"] = {k: int(v.sum()) for k, v in ph_flags.items()}

# ============================================================================
# One-table summary of both datasets
def summarise(name, df, res, time_info, key_note):
    return {
        "Dataset": name, "Observations": res["n_obs"], "Variables": res["n_vars"],
        "Numeric variables": len(res["numeric_cols"]), "Missing cells": res["missing_cells"],
        "Duplicate rows": res["n_duplicate_rows"], "Infinite values": res["n_inf"],
        "Time information": time_info, "Role": key_note,
    }
dataset_summary = pd.DataFrame([
    summarise("wage1", wage1, audit_results["wage1"], "none (cross-section)", "Primary: wage determination"),
    summarise("phillips", phillips, audit_results["phillips"],
              f"annual, {yr.min():.0f}-{yr.max():.0f}" if phillips_time_ok else "needs verification",
              "Supplementary: autocorrelation (Part 5)"),
]).set_index("Dataset")
display(dataset_summary.T)

# ============================================================================
# Formal preparation checks on the working copies
def numeric_integrity(df, name):
    rows = []
    for c in df.columns:
        coerced = pd.to_numeric(df[c], errors="coerce")
        rows.append({
            "variable": c,
            "numeric_dtype": pd.api.types.is_numeric_dtype(df[c]),
            "entries_not_convertible_to_number": int(coerced.isna().sum() - df[c].isna().sum()),
            "infinite_values": int(np.isinf(coerced).sum()),
            "missing_values": int(df[c].isna().sum()),
        })
    out = pd.DataFrame(rows).set_index("variable")
    print(f"{name}: numeric_dtype=False for {int((~out['numeric_dtype']).sum())} columns; "
          f"non-convertible entries={int(out['entries_not_convertible_to_number'].sum())}; "
          f"infinite values={int(out['infinite_values'].sum())}; "
          f"missing values={int(out['missing_values'].sum())}")
    return out

integrity_wage1 = numeric_integrity(wage1, "wage1")
integrity_phillips = numeric_integrity(phillips, "phillips")
print("\n(Full per-variable tables are stored in integrity_wage1 and integrity_phillips.)")

section("wage1: inspection of the main wage variable")
wage = wage1["wage"]
q1, q3 = wage.quantile([0.25, 0.75])
tukey_upper = q3 + 1.5 * (q3 - q1)
wage_profile = pd.Series({
    "count": wage.count(), "min": wage.min(), "1st percentile": wage.quantile(0.01),
    "median": wage.median(), "99th percentile": wage.quantile(0.99), "max": wage.max(),
    "values <= 0": int((wage <= 0).sum()), "values missing": int(wage.isna().sum()),
    "values above Tukey upper fence (Q3 + 1.5 IQR)": int((wage > tukey_upper).sum()),
}, name="wage")
display(wage_profile.to_frame())
print(f"Tukey upper fence = {tukey_upper:.3f}. Values above it are common in right-skewed pay data "
      "and are only flagged for inspection, NOT removed.")
print("\nFive lowest and five highest wages (with the worker's other recorded characteristics):")
cols_show = ["wage", "educ", "exper", "tenure", "female", "married", "nonwhite"]
display(wage1.nsmallest(5, "wage")[cols_show])
display(wage1.nlargest(5, "wage")[cols_show])

# ============================================================================
# Decisions are generated from the audit results above, then printed for the record
n_w_miss = audit_results["wage1"]["missing_cells"]
n_p_miss = audit_results["phillips"]["missing_cells"]
n_nonpos = int((wage1["wage"] <= 0).sum())

decisions = pd.DataFrame([
    ("Row deletion (both datasets)", "No observations are removed. Unusual values are flagged, not treated as errors."),
    ("Outliers in wage", "Retained. High wages are plausible in survey data; any robustness treatment is deferred and would be justified explicitly."),
    ("Missing values - wage1", f"{n_w_miss} missing cells found. " + ("No imputation needed." if n_w_miss == 0 else "Not imputed; decide before Part 2.")),
    ("Missing values - phillips", f"{n_p_miss} missing cells found" + (", all in lagged/differenced variables for the first year (by construction). Not imputed; Part 5 will use the lags it actually needs." if n_p_miss else ".")),
    ("Duplicate rows", f"wage1 has {audit_results['wage1']['n_duplicate_rows']} and phillips has {audit_results['phillips']['n_duplicate_rows']} exact duplicate rows. Not deleted (no ID variable to establish true duplication)."),
    ("New variable: log_wage", f"Created as natural log of wage for strictly positive wages. Observations with wage <= 0 would receive NaN (currently {n_nonpos}). Original 'wage' and 'lwage' are preserved."),
    ("Existing transformed variables", "lwage, expersq and tenursq (supplied by the dataset) are kept unchanged; they are verified against the originals in Cell 9."),
], columns=["Item", "Decision"]).set_index("Item")
with pd.option_context("display.max_colwidth", None):
    display(decisions)

# ============================================================================
# Create log_wage in the WORKING copy only (the raw copy is untouched)
n_before = len(wage1)
positive = wage1["wage"] > 0
n_cannot_log = int((~positive).sum())          # wage <= 0 (log undefined)
n_missing_wage = int(wage1["wage"].isna().sum())

wage1["log_wage"] = np.where(positive, np.log(wage1["wage"].where(positive)), np.nan)

print(f"Observations before transformation : {n_before}")
print(f"Observations after transformation  : {len(wage1)}  (no rows dropped)")
print(f"Observations that cannot be logged : {n_cannot_log} with wage <= 0 (log undefined)")
print(f"Observations with missing wage     : {n_missing_wage}")
print(f"log_wage non-missing observations  : {int(wage1['log_wage'].notna().sum())}")

# Cross-check with the dataset's own 'lwage' variable
max_diff = float((wage1["log_wage"] - wage1["lwage"]).abs().max(skipna=True))
print(f"Largest absolute difference between log_wage and the supplied lwage: {max_diff:.2e}")

NEW_VARIABLES = {"log_wage": "natural log of wage, created in Part 1 (NaN where wage <= 0 or missing)"}
print("\nNew variables created in Part 1:", NEW_VARIABLES)
print("Raw data unchanged and working copy has one extra column:",
      list(wage1.columns.difference(wage1_raw.columns)) == ["log_wage"])

# ============================================================================
# Data-quality report (all statements are computed from the data above)
w_res, p_res = audit_results["wage1"], audit_results["phillips"]
w_issues, w_notes = [], []
if w_res["missing_cells"] > 0:               w_issues.append(f"{w_res['missing_cells']} missing cells")
if w_res["n_inf"] > 0:                       w_issues.append(f"{w_res['n_inf']} infinite values")
if w_res["non_numeric_cols"]:                w_issues.append(f"non-numeric columns: {w_res['non_numeric_cols']}")
if absent:                                   w_issues.append(f"expected variables absent: {absent}")
if n_cannot_log > 0:                         w_issues.append(f"{n_cannot_log} wages cannot be logged")
for k, v in w_res["flag_counts"].items():
    if v > 0: w_notes.append(f"{v} rows flagged: {k}")
if w_res["n_duplicate_rows"] > 0:            w_notes.append(f"{w_res['n_duplicate_rows']} exact duplicate rows retained (no ID variable)")

p_issues, p_notes = [], []
if p_res["n_inf"] > 0:                       p_issues.append(f"{p_res['n_inf']} infinite values")
if p_res["non_numeric_cols"]:                p_issues.append(f"non-numeric columns: {p_res['non_numeric_cols']}")
if not phillips_time_ok:                     p_issues.append("year does not form a clean annual sequence")
if not phillips_defs_ok:                     p_issues.append("derived variables disagree with documented definitions")
if p_res["missing_cells"] > 0:               p_notes.append(f"{p_res['missing_cells']} missing cells (first-year lags/differences, by construction)")
for k, v in p_res["flag_counts"].items():
    if v > 0: p_notes.append(f"{v} rows flagged: {k}")

print("DATA-QUALITY REPORT")
print("=" * 80)
for nm, n0, issues, notes in [("wage1", initial_n["wage1"], w_issues, w_notes),
                              ("phillips", initial_n["phillips"], p_issues, p_notes)]:
    print(f"\n{nm}  (initial sample size: {n0})")
    print("  Blocking issues :", issues if issues else "none found")
    print("  Notes           :", notes if notes else "none")
    print("  Status          :", "READY for subsequent analysis" if not issues else "NEEDS REVIEW before Part 2/5")
data_quality = {"wage1_ready": not w_issues, "phillips_ready": not p_issues}

# ============================================================================
# Descriptive statistics table (count, mean, median, std, min, max)
core_vars = ["wage", "educ", "exper", "tenure"]
dummy_vars = [c for c in ["female", "married", "nonwhite"] if c in wage1.columns]
eda_vars = core_vars + dummy_vars

def describe_table(df, cols):
    t = df[cols].agg(["count", "mean", "median", "std", "min", "max"]).T
    t.insert(0, "documented meaning", [WAGE1_DOC.get(c, "") for c in cols])
    return t

desc_wage1 = describe_table(wage1, eda_vars)
with pd.option_context("display.max_colwidth", None):
    display(desc_wage1)

# ============================================================================
# Histogram with density, boxplot, and histogram of log wages
x = wage1["wage"].dropna()
logw = wage1["log_wage"].dropna()

fig, axes = plt.subplots(1, 3, figsize=(16, 4.6))

# (a) histogram + kernel density
ax = axes[0]
ax.hist(x, bins=30, density=True, color=BLUE, edgecolor="white", alpha=0.85, label="Histogram")
grid = np.linspace(x.min(), x.max(), 400)
ax.plot(grid, stats.gaussian_kde(x)(grid), color=RED, lw=2, label="Kernel density")
ax.axvline(x.mean(), color="black", ls="--", lw=1.3, label=f"Mean = {x.mean():.2f}")
ax.axvline(x.median(), color=GREY, ls=":", lw=1.8, label=f"Median = {x.median():.2f}")
ax.set(title="(a) Distribution of hourly wage", xlabel="Average hourly earnings", ylabel="Density")
ax.legend()

# (b) boxplot
ax = axes[1]
sns.boxplot(x=x, ax=ax, color=BLUE, width=0.45, fliersize=3)
ax.set(title="(b) Boxplot of hourly wage", xlabel="Average hourly earnings")

# (c) log wage histogram (only strictly positive wages are defined)
ax = axes[2]
if n_cannot_log == 0:
    note = f"All {len(logw)} wages are strictly positive; none excluded."
else:
    note = f"{n_cannot_log} observations with wage <= 0 are excluded (log undefined)."
ax.hist(logw, bins=30, density=True, color=BLUE, edgecolor="white", alpha=0.85, label="Histogram")
grid2 = np.linspace(logw.min(), logw.max(), 400)
ax.plot(grid2, stats.gaussian_kde(logw)(grid2), color=RED, lw=2, label="Kernel density")
ax.set(title="(c) Distribution of log hourly wage", xlabel="log(wage)", ylabel="Density")
ax.legend()

fig.suptitle("wage1: distribution of hourly earnings", fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
save_fig(fig, "fig1_wage_distribution")
plt.show()

print(note)
print(f"Skewness of wage     : {stats.skew(x):.3f}")
print(f"Skewness of log_wage : {stats.skew(logw):.3f}")

# ============================================================================
# Scatterplots with exploratory trend lines
def scatter_with_trend(ax, xvar, xlabel, title):
    d = wage1[[xvar, "wage"]].dropna()
    ax.scatter(d[xvar], d["wage"], s=18, alpha=0.40, color=BLUE, edgecolor="none")
    # straight-line trend (drawn only for visual guidance; coefficients are not reported)
    slope, intercept = np.polyfit(d[xvar], d["wage"], 1)
    xs = np.linspace(d[xvar].min(), d[xvar].max(), 100)
    ax.plot(xs, intercept + slope * xs, color=RED, lw=2, ls="--", label="Linear trend (exploratory)")
    # LOWESS: a flexible smoother that shows whether the pattern is curved
    smooth = sm.nonparametric.lowess(d["wage"], d[xvar], frac=0.6)
    ax.plot(smooth[:, 0], smooth[:, 1], color="black", lw=1.8, label="LOWESS smoother (exploratory)")
    ax.set(title=title, xlabel=xlabel, ylabel="Average hourly earnings")

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))
scatter_with_trend(axes[0], "educ", "Years of education", "(a) Wage vs education")
scatter_with_trend(axes[1], "exper", "Years of potential experience", "(b) Wage vs experience")
scatter_with_trend(axes[2], "tenure", "Years with current employer", "(c) Wage vs tenure")
h, l = axes[0].get_legend_handles_labels()
fig.legend(h, l, loc="lower center", ncol=2, frameon=False)
fig.suptitle("wage1: wage against education, experience and tenure (descriptive, not causal)",
             fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout(rect=[0, 0.07, 1, 1])
save_fig(fig, "fig2_wage_scatterplots")
plt.show()

# ============================================================================
# Confirm coding, then compare wages across groups (descriptive only)
print("Coding according to the dataset documentation:")
for c in ["female", "married"]:
    print(f"  {c:8s}: {WAGE1_DOC[c]}   | observed values in data: {sorted(wage1[c].dropna().unique().tolist())}")
assert set(wage1["female"].dropna().unique()) <= {0, 1} and set(wage1["married"].dropna().unique()) <= {0, 1}

# Labelled helper copy used for plotting/tables only; wage1 itself is not modified
plot_df = wage1.copy()
plot_df["Gender"] = plot_df["female"].map({0: "Male\n(female=0)", 1: "Female\n(female=1)"})
plot_df["Marital status"] = plot_df["married"].map({0: "Not married\n(married=0)", 1: "Married\n(married=1)"})
plot_df["Group"] = plot_df["Gender"].str.split("\n").str[0] + ", " + \
                   plot_df["Marital status"].str.split("\n").str[0].str.lower()

mean_marker = dict(marker="D", markerfacecolor="white", markeredgecolor="black", markersize=6)
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
for ax, col, ttl in [(axes[0], "Gender", "(a) By gender"), (axes[1], "Marital status", "(b) By marital status")]:
    order = sorted(plot_df[col].dropna().unique())
    sns.boxplot(data=plot_df, x=col, y="wage", order=order, ax=ax, color=BLUE, width=0.5,
                showmeans=True, meanprops=mean_marker, fliersize=3)
    ax.set(title=ttl, xlabel="", ylabel="Average hourly earnings")
order4 = sorted(plot_df["Group"].dropna().unique())
sns.boxplot(data=plot_df, x="Group", y="wage", order=order4, ax=axes[2], color=BLUE, width=0.5,
            showmeans=True, meanprops=mean_marker, fliersize=3)
axes[2].set(title="(c) By gender and marital status", xlabel="", ylabel="Average hourly earnings")
axes[2].tick_params(axis="x", labelsize=8)
fig.suptitle("wage1: wage distributions by group (diamond = group mean; descriptive only)",
             fontsize=13, fontweight="bold", y=1.02)
plt.tight_layout()
save_fig(fig, "fig3_wage_by_groups")
plt.show()

def group_summary(by):
    t = plot_df.groupby(by, observed=True)["wage"].agg(["count", "mean", "median", "std", "min", "max"])
    return t

for by in ["female", "married", ["female", "married"]]:
    print(f"\nWage summary by {by}:")
    display(group_summary(by))

print("\nAverage characteristics of each group (shows that groups differ in more than one way):")
display(plot_df.groupby(["female", "married"])[["educ", "exper", "tenure"]].mean())

# ============================================================================
# Pearson correlation matrix and heatmap
corr_vars = ["wage", "log_wage", "educ", "exper", "tenure"] + dummy_vars
corr = wage1[corr_vars].corr(method="pearson")
print("Pearson correlation matrix:")
display(corr)

mask = np.triu(np.ones_like(corr, dtype=bool), k=1)   # hide the duplicate upper triangle
fig, ax = plt.subplots(figsize=(8.5, 6.5))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1, center=0,
            square=True, linewidths=0.5, cbar_kws={"label": "Pearson correlation"}, ax=ax)
ax.set_title("wage1: correlation matrix of key variables")
plt.tight_layout()
save_fig(fig, "fig4_correlation_heatmap")
plt.show()

# ============================================================================
# Descriptive statistics for phillips
ph_vars = ["unem", "inf", "cunem", "cinf"]
ph_desc = phillips[ph_vars].agg(["count", "mean", "median", "std", "min", "max"]).T
ph_desc.insert(0, "documented meaning", [PHILLIPS_DOC[c] for c in ph_vars])
with pd.option_context("display.max_colwidth", None):
    display(ph_desc)

time_candidates = [c for c in phillips.columns if "year" in c.lower() or "date" in c.lower()
                   or "time" in c.lower() or "period" in c.lower()]
print("Candidate time variables:", time_candidates)
print("Time index established (documented 'year', ordered, gap-free annual sequence):", phillips_time_ok)

# ============================================================================
# Plot phillips variables over time (conditional on a verified time index)
if not phillips_time_ok:
    print("Time plots omitted because the annual time index failed validation.")
else:
    fig, axes = plt.subplots(2, 2, figsize=(14, 9))

    ax = axes[0, 0]
    ax.plot(phillips["year"], phillips["unem"], color=BLUE, lw=1.8, marker="o", ms=3)
    ax.set(title="(a) Civilian unemployment rate", xlabel="Year", ylabel="Unemployment rate, %")

    ax = axes[0, 1]
    ax.plot(phillips["year"], phillips["inf"], color=RED, lw=1.8, marker="o", ms=3)
    ax.axhline(0, color=GREY, lw=0.8)
    ax.set(title="(b) Inflation (percentage change in CPI)", xlabel="Year", ylabel="Inflation, %")

    ax = axes[1, 0]
    ax.plot(phillips["year"], phillips["cunem"], color=BLUE, lw=1.5, label="cunem = unem - unem_1")
    ax.plot(phillips["year"], phillips["cinf"], color=RED, lw=1.5, label="cinf = inf - inf_1")
    ax.axhline(0, color=GREY, lw=0.8)
    ax.set(title="(c) Year-to-year changes", xlabel="Year", ylabel="Change in percentage points")
    ax.legend()

    ax = axes[1, 1]
    ax.scatter(phillips["unem"], phillips["inf"], s=28, color=BLUE, alpha=0.8, edgecolor="none")
    ax.set(title="(d) Inflation vs unemployment (same-year, no trend line)",
           xlabel="Unemployment rate, %", ylabel="Inflation, %")

    fig.suptitle("phillips: unemployment and inflation, annual data", fontsize=13, fontweight="bold", y=1.01)
    plt.tight_layout()
    save_fig(fig, "fig5_phillips_time_plots")
    plt.show()


# ============================================================================
# What must be verified/used before Part 5
print(f"  - Time variable: 'year', {yr.min():.0f}-{yr.max():.0f}, {len(phillips)} annual observations, "
      f"ordered and gap-free: {phillips_time_ok}")
print(f"  - Derived variables match documented definitions: {phillips_defs_ok}")
print(f"  - Complete (no-missing) rows available: {int(phillips.dropna().shape[0])} of {len(phillips)}")

# ============================================================================
# Final summary
# Re-check that the raw copies were never modified
raw_untouched = (
    int(pd.util.hash_pandas_object(wage1_raw, index=True).sum()) == raw_fingerprint["wage1_raw"] and
    int(pd.util.hash_pandas_object(phillips_raw, index=True).sum()) == raw_fingerprint["phillips_raw"]
)

print("=" * 80)
print("PART 1 SUMMARY")
print("=" * 80)
print(f"Loaded via the wooldridge package : wage1 ({wage1_raw.shape[0]} obs x {wage1_raw.shape[1]} vars), "
      f"phillips ({phillips_raw.shape[0]} obs x {phillips_raw.shape[1]} vars)")
print(f"Initial sample sizes recorded     : {initial_n}")
print(f"Raw copies unchanged              : {raw_untouched}")
print(f"Observations removed              : none (wage1: {len(wage1)} of {initial_n['wage1']}, "
      f"phillips: {len(phillips)} of {initial_n['phillips']})")
print(f"New variable created              : log_wage (wage1); {n_cannot_log} observation(s) could not be logged")
print(f"Expected wage1 variables present  : {not absent}")
print(f"wage1 data-quality status         : {'ready' if data_quality['wage1_ready'] else 'needs review'}")
print(f"phillips data-quality status      : {'ready' if data_quality['phillips_ready'] else 'needs review'}")
print("\nData-quality notes:")
for nm, notes in [("wage1", w_notes), ("phillips", p_notes)]:
    for n_ in (notes if notes else ["none"]):
        print(f"  [{nm}] {n_}")
print(f"\nFigures saved as PNG in: {FIG_DIR.resolve()}")
can_proceed = raw_untouched and data_quality["wage1_ready"] and data_quality["phillips_ready"] and not absent


# ============================================================================
# Export Part 1 results for the final project report

from pathlib import Path
import json

OUTPUT_DIR = Path("/content/outputs_part1")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Main descriptive and data-quality tables
tables = {
    "dataset_summary": dataset_summary,
    "wage1_descriptive_statistics": desc_wage1,
    "phillips_descriptive_statistics": ph_desc,
    "wage1_correlation_matrix": corr,
    "wage1_plausibility_flags": flag_table,
    "data_preparation_decisions": decisions,
    "wage1_numeric_integrity": integrity_wage1,
    "phillips_numeric_integrity": integrity_phillips,
    "phillips_derived_variable_checks": phillips_checks.to_frame(),
    "wage_profile": wage_profile.to_frame(),
    "wage1_group_gender": group_summary("female"),
    "wage1_group_marital_status": group_summary("married"),
    "wage1_group_gender_marital": group_summary(["female", "married"]),
    "wage1_group_characteristics": plot_df.groupby(
        ["female", "married"]
    )[["educ", "exper", "tenure"]].mean(),
    "wage1_indicator_frequencies": pd.DataFrame(freq_rows),
    "wage1_variable_presence": presence,
    "phillips_time_series_checks": pd.DataFrame({
        "check": [
            "year_whole_numbers",
            "year_sorted",
            "year_unique",
            "year_consecutive",
            "derived_variables_consistent"
        ],
        "passed": [
            yr_whole,
            yr_sorted,
            yr_unique,
            yr_consecutive,
            phillips_defs_ok
        ]
    }),
}

for name, table in tables.items():
    table.to_csv(OUTPUT_DIR / f"{name}.csv", index=True)

# Export audit summaries
for dataset_name, result in audit_results.items():
    audit_table = pd.DataFrame(
        [{"metric": key, "value": str(value)}
         for key, value in result.items()]
    )
    audit_table.to_csv(
        OUTPUT_DIR / f"{dataset_name}_audit_summary.csv",
        index=False
    )

# Save the full data-quality status
with open(OUTPUT_DIR / "data_quality_status.json", "w") as f:
    json.dump(data_quality, f, indent=2)

# Save the dataset copies used in the project
wage1.to_csv(OUTPUT_DIR / "wage1_working_data.csv", index=False)
phillips.to_csv(OUTPUT_DIR / "phillips_working_data.csv", index=False)

print(f"\nPart 1 tables and data exported to: {OUTPUT_DIR}")
print(f"CSV tables exported: {len(tables) + len(audit_results)}")
print("Working dataset copies exported: wage1_working_data.csv, phillips_working_data.csv")

```

```python
# Part 2: OLS Regression and Statistical Inference

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy import stats
from IPython.display import display
from pathlib import Path

# ---------------------------------------------------------------------------
# Output directory
OUTPUT_DIR_PART2 = Path("outputs_part2")
OUTPUT_DIR_PART2.mkdir(parents=True, exist_ok=True)

TABLE_DIR_PART2 = OUTPUT_DIR_PART2 / "tables"
TABLE_DIR_PART2.mkdir(parents=True, exist_ok=True)

pd.set_option("display.max_columns", 30)
pd.set_option("display.width", 140)
pd.set_option("display.float_format", lambda x: f"{x:,.4f}")

# ---------------------------------------------------------------------------
# Confirm that Part 1 has been run and the required dataset is available
if "wage1" not in globals():
    raise NameError(
        "The DataFrame 'wage1' was not found. Run Part 1 first in the same "
        "Google Colab notebook, then run this script."
    )

if not isinstance(wage1, pd.DataFrame):
    raise TypeError("'wage1' must be a pandas DataFrame.")

required_variables = [
    "wage", "educ", "exper", "tenure", "female", "married", "nonwhite"
]
missing_variables = [variable for variable in required_variables if variable not in wage1.columns]

if missing_variables:
    raise ValueError(
        f"The wage1 dataset is missing required variables: {missing_variables}"
    )

# Preserve the Part 1 dataset and create a model-specific working copy.
wage1_part2 = wage1.copy(deep=True)

# Convert selected variables to numeric only in the working copy.
for variable in required_variables:
    wage1_part2[variable] = pd.to_numeric(
        wage1_part2[variable], errors="coerce"
    )

# Replace infinite values with missing values in the model-specific copy.
wage1_part2[required_variables] = wage1_part2[required_variables].replace(
    [np.inf, -np.inf], np.nan
)

# ---------------------------------------------------------------------------
# Sample definition and descriptive summary
model_variables = required_variables
model_data = wage1_part2[model_variables].dropna().copy()

if model_data.empty:
    raise ValueError("No complete observations remain for the specified models.")

if (model_data["wage"] <= 0).any():
    raise ValueError(
        "The selected wage sample contains non-positive wages. Inspect the data "
        "before estimating the models."
    )

sample_summary = pd.DataFrame(
    {
        "Initial observations": [len(wage1_part2)],
        "Complete observations": [len(model_data)],
        "Observations excluded for missing/non-finite values": [
            len(wage1_part2) - len(model_data)
        ],
    },
    index=["wage1 model sample"],
)
display(sample_summary)
sample_summary.to_csv(TABLE_DIR_PART2 / "sample_summary.csv")

descriptive_statistics = model_data.describe().T
descriptive_statistics["median"] = model_data.median()
descriptive_statistics = descriptive_statistics[
    ["count", "mean", "median", "std", "min", "max"]
]
display(descriptive_statistics)
descriptive_statistics.to_csv(TABLE_DIR_PART2 / "model_sample_descriptive_statistics.csv")

# ---------------------------------------------------------------------------
# Model 1: Simple OLS regression of hourly wage on education
simple_data = model_data[["wage", "educ"]].dropna().copy()
X_simple = sm.add_constant(simple_data[["educ"]], has_constant="add")
y_simple = simple_data["wage"]

model_simple = sm.OLS(y_simple, X_simple).fit()

print("Model 1: Simple OLS regression")
display(model_simple.summary())

# ---------------------------------------------------------------------------
# Model 2: Multiple OLS regression with worker characteristics
multiple_formula = "wage ~ educ + exper + tenure + female + married + nonwhite"
model_multiple = smf.ols(formula=multiple_formula, data=model_data).fit()

print("Model 2: Multiple OLS regression")
display(model_multiple.summary())

# ---------------------------------------------------------------------------
# Comparable regression table
def build_coefficient_table(models, model_names):
    rows = []
    for model, model_name in zip(models, model_names):
        confidence_intervals = model.conf_int()
        for term in model.params.index:
            rows.append(
                {
                    "Model": model_name,
                    "Variable": term,
                    "Coefficient": model.params[term],
                    "Standard error": model.bse[term],
                    "t-statistic": model.tvalues[term],
                    "p-value": model.pvalues[term],
                    "95% CI lower": confidence_intervals.loc[term, 0],
                    "95% CI upper": confidence_intervals.loc[term, 1],
                }
            )
    return pd.DataFrame(rows)

coefficient_table = build_coefficient_table(
    [model_simple, model_multiple],
    ["Simple OLS", "Multiple OLS"],
)
display(coefficient_table)
coefficient_table.to_csv(
    TABLE_DIR_PART2 / "ols_coefficient_comparison.csv", index=False
)

model_fit_table = pd.DataFrame(
    {
        "Model": ["Simple OLS", "Multiple OLS"],
        "Dependent variable": ["wage", "wage"],
        "Observations": [int(model_simple.nobs), int(model_multiple.nobs)],
        "R-squared": [model_simple.rsquared, model_multiple.rsquared],
        "Adjusted R-squared": [
            model_simple.rsquared_adj,
            model_multiple.rsquared_adj,
        ],
        "Residual standard error": [
            np.sqrt(model_simple.mse_resid),
            np.sqrt(model_multiple.mse_resid),
        ],
        "F-statistic": [model_simple.fvalue, model_multiple.fvalue],
        "F-test p-value": [model_simple.f_pvalue, model_multiple.f_pvalue],
    }
)
display(model_fit_table)
model_fit_table.to_csv(TABLE_DIR_PART2 / "ols_model_fit_comparison.csv", index=False)

# ---------------------------------------------------------------------------
# Individual coefficient t-tests
# The conventional OLS t-tests use the standard errors estimated by each model.
def individual_t_test_table(model, model_name):
    ci = model.conf_int()
    return pd.DataFrame(
        {
            "Model": model_name,
            "Coefficient": model.params,
            "Standard error": model.bse,
            "t-statistic": model.tvalues,
            "p-value": model.pvalues,
            "95% CI lower": ci[0],
            "95% CI upper": ci[1],
            "Reject H0 at 5%": model.pvalues < 0.05,
        }
    ).rename_axis("Variable").reset_index()

t_test_results = pd.concat(
    [
        individual_t_test_table(model_simple, "Simple OLS"),
        individual_t_test_table(model_multiple, "Multiple OLS"),
    ],
    ignore_index=True,
)
display(t_test_results)
t_test_results.to_csv(TABLE_DIR_PART2 / "individual_t_tests.csv", index=False)

# ---------------------------------------------------------------------------
# Overall F-tests for joint significance of all non-constant regressors
def overall_f_test_table(model, model_name):
    return {
        "Model": model_name,
        "Null hypothesis": "All slope coefficients equal zero",
        "F-statistic": model.fvalue,
        "Numerator df": model.df_model,
        "Denominator df": model.df_resid,
        "p-value": model.f_pvalue,
        "Reject H0 at 5%": bool(model.f_pvalue < 0.05),
    }

overall_f_results = pd.DataFrame(
    [
        overall_f_test_table(model_simple, "Simple OLS"),
        overall_f_test_table(model_multiple, "Multiple OLS"),
    ]
)
display(overall_f_results)
overall_f_results.to_csv(
    TABLE_DIR_PART2 / "overall_f_tests.csv", index=False
)

# ---------------------------------------------------------------------------
# Joint F-test for experience and tenure in the multiple regression
joint_restriction = model_multiple.f_test("exper = 0, tenure = 0")
joint_f_result = pd.DataFrame(
    [
        {
            "Test": "Joint significance of experience and tenure",
            "Null hypothesis": "beta_exper = 0 and beta_tenure = 0",
            "F-statistic": float(np.asarray(joint_restriction.fvalue).squeeze()),
            "Numerator df": int(np.asarray(joint_restriction.df_num).squeeze()),
            "Denominator df": float(np.asarray(joint_restriction.df_denom).squeeze()),
            "p-value": float(np.asarray(joint_restriction.pvalue).squeeze()),
            "Reject H0 at 5%": bool(float(np.asarray(joint_restriction.pvalue).squeeze()) < 0.05),
        }
    ]
)
display(joint_f_result)
joint_f_result.to_csv(TABLE_DIR_PART2 / "joint_f_test_experience_tenure.csv", index=False)

# ---------------------------------------------------------------------------
# Compact coefficient interpretation for the multiple regression
interpretation_rows = []
for variable in ["educ", "exper", "tenure", "female", "married", "nonwhite"]:
    coefficient = float(model_multiple.params[variable])
    p_value = float(model_multiple.pvalues[variable])
    interpretation_rows.append(
        {
            "Variable": variable,
            "Estimated coefficient": coefficient,
            "p-value": p_value,
            "Statistical significance at 5%": (
                "Statistically significant" if p_value < 0.05
                else "Not statistically significant"
            ),
            "Interpretation": (
                f"Holding the other included regressors constant, a one-unit "
                f"increase in {variable} is associated with an estimated "
                f"{coefficient:.4f}-unit change in hourly wage. This is an "
                f"association, not necessarily a causal effect."
            ),
        }
    )

interpretation_table = pd.DataFrame(interpretation_rows)
display(interpretation_table)
interpretation_table.to_csv(
    TABLE_DIR_PART2 / "multiple_model_interpretation.csv", index=False
)

# ---------------------------------------------------------------------------
# Save model objects for use in later parts of the same notebook session.
# The fitted objects remain available as model_simple and model_multiple.
print("Part 2 outputs saved to:", OUTPUT_DIR_PART2.resolve())
print("Estimated models available: model_simple, model_multiple")
```

```python
# Part 3 — Functional Forms and Dummy Variables

from pathlib import Path
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from IPython.display import display

OUT = Path('outputs_part3')
TABLES = OUT / 'tables'
TABLES.mkdir(parents=True, exist_ok=True)

required = ['wage', 'educ', 'exper', 'tenure', 'female', 'married', 'nonwhite']
if 'wage1' not in globals() or not isinstance(wage1, pd.DataFrame):
    raise RuntimeError('Run Part 1 in this Colab session before Part 3.')
missing = sorted(set(required) - set(wage1.columns))
if missing:
    raise ValueError(f'Required wage1 columns missing: {missing}')

# Common sample across all specifications
part3_data = wage1[required].copy()
for col in required:
    part3_data[col] = pd.to_numeric(part3_data[col], errors='coerce')
part3_data = part3_data.replace([np.inf, -np.inf], np.nan).dropna().copy()
part3_data = part3_data.loc[part3_data['wage'] > 0].copy()
if len(part3_data) < 30:
    raise ValueError('Insufficient complete, positive-wage observations.')
part3_data['log_wage'] = np.log(part3_data['wage'])

# Base specifications, using exactly the same observations and regressors
rhs = 'educ + exper + tenure + female + married + nonwhite'
part3_linear = smf.ols(f'wage ~ {rhs}', data=part3_data).fit()
part3_loglinear = smf.ols(f'log_wage ~ {rhs}', data=part3_data).fit()

# Coefficient table (do not compare R-squared across different dependent variables)
def coefficient_table(models):
    rows = []
    for name, fitted in models.items():
        ci = fitted.conf_int()
        for term in fitted.params.index:
            rows.append({
                'model': name, 'term': term,
                'coefficient': fitted.params[term],
                'std_error': fitted.bse[term],
                't_statistic': fitted.tvalues[term],
                'p_value': fitted.pvalues[term],
                'ci_lower_95': ci.loc[term, 0],
                'ci_upper_95': ci.loc[term, 1],
            })
    return pd.DataFrame(rows)

part3_models = {'Linear wage': part3_linear, 'Log wage': part3_loglinear}
part3_coefficients = coefficient_table(part3_models)
display(part3_coefficients.round(5))
part3_coefficients.to_csv(TABLES / 'functional_form_coefficients.csv', index=False)

part3_model_summary = pd.DataFrame([
    {'model': name, 'dependent_variable': model.model.endog_names,
     'n': int(model.nobs), 'r_squared': model.rsquared,
     'adjusted_r_squared': model.rsquared_adj, 'aic': model.aic}
    for name, model in part3_models.items()
])
display(part3_model_summary.round(5))
part3_model_summary.to_csv(TABLES / 'functional_form_model_summary.csv', index=False)

# Marginal associations in the log-wage model:
# continuous regressors: 100*(exp(beta)-1) % change for a one-unit increase;
# binary regressors: 100*(exp(beta)-1) % difference between indicator groups.
part3_log_interpretation = pd.DataFrame([
    {'variable': term, 'coefficient': part3_loglinear.params[term],
     'implied_percent_difference': 100 * np.expm1(part3_loglinear.params[term]),
     'indicator': term in ['female', 'married', 'nonwhite']}
    for term in required if term != 'wage'
])
display(part3_log_interpretation.round(4))
part3_log_interpretation.to_csv(TABLES / 'log_wage_coefficient_interpretation.csv', index=False)

# Dummy-variable joint F-test in the log-wage specification
part3_dummy_test = part3_loglinear.f_test('female = 0, married = 0, nonwhite = 0')
part3_dummy_results = pd.DataFrame([{
    'test': 'Joint F-test: worker-characteristic indicators',
    'null': 'female = married = nonwhite = 0',
    'f_statistic': float(np.asarray(part3_dummy_test.fvalue).item()),
    'df_num': float(part3_dummy_test.df_num),
    'df_denom': float(part3_dummy_test.df_denom),
    'p_value': float(np.asarray(part3_dummy_test.pvalue).item()),
}])
part3_dummy_results['reject_at_5pct'] = part3_dummy_results['p_value'] < .05
display(part3_dummy_results)
part3_dummy_results.to_csv(TABLES / 'joint_dummy_f_test.csv', index=False)

# Interaction extension: allow the education slope to differ by the female indicator
part3_interaction = smf.ols(
    'log_wage ~ educ + exper + tenure + female + married + nonwhite + educ:female',
    data=part3_data
).fit()
part3_interaction_table = coefficient_table({'Log wage with education interaction': part3_interaction})
display(part3_interaction_table.round(5))
part3_interaction_table.to_csv(TABLES / 'education_indicator_interaction.csv', index=False)

# MacKinnon–White–Davidson (MWD) non-nested functional-form comparison.
# Step 1: Estimate wage and log(wage) models using identical regressors and sample.
# Step 2: Z1 = log(fitted wage from linear model) - fitted log wage.
#         Z2 = exp(fitted log wage) - fitted wage.
# Step 3: Add Z1 to the log-wage equation and Z2 to the wage equation.
#         The t-test of each auxiliary coefficient tests that equation's adequacy.
# The log(fitted wage) calculation requires strictly positive linear predictions.
linear_pred = np.asarray(part3_linear.fittedvalues)
log_pred = np.asarray(part3_loglinear.fittedvalues)
if np.all(np.isfinite(linear_pred)) and np.all(linear_pred > 0):
    mwd_data = part3_data.copy()
    mwd_data['mwd_z1'] = np.log(linear_pred) - log_pred
    mwd_data['mwd_z2'] = np.exp(log_pred) - linear_pred
    part3_mwd_log = smf.ols(f'log_wage ~ {rhs} + mwd_z1', data=mwd_data).fit()
    part3_mwd_linear = smf.ols(f'wage ~ {rhs} + mwd_z2', data=mwd_data).fit()
    part3_mwd_results = pd.DataFrame([
        {'null_model': 'Log wage', 'auxiliary_term': 'mwd_z1',
         'coefficient': part3_mwd_log.params['mwd_z1'],
         't_statistic': part3_mwd_log.tvalues['mwd_z1'],
         'p_value': part3_mwd_log.pvalues['mwd_z1']},
        {'null_model': 'Linear wage', 'auxiliary_term': 'mwd_z2',
         'coefficient': part3_mwd_linear.params['mwd_z2'],
         't_statistic': part3_mwd_linear.tvalues['mwd_z2'],
         'p_value': part3_mwd_linear.pvalues['mwd_z2']},
    ])
    part3_mwd_results['reject_at_5pct'] = part3_mwd_results['p_value'] < .05
    display(part3_mwd_results.round(6))
    part3_mwd_results.to_csv(TABLES / 'mwd_functional_form_test.csv', index=False)
else:
    part3_mwd_results = pd.DataFrame([{
        'status': 'Not computed',
        'reason': 'Nonpositive or nonfinite fitted wages prevent log(fitted wage).'
    }])
    display(part3_mwd_results)
    part3_mwd_results.to_csv(TABLES / 'mwd_functional_form_test.csv', index=False)

# Retain fitted models and common sample for subsequent analysis.
```

```python
# Part 4: Classical Regression Diagnostics

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.stats.diagnostic import het_breuschpagan, het_white
from IPython.display import display
from pathlib import Path

# ---------------------------------------------------------------------------
# Output directories
OUTPUT_DIR_PART4 = Path("outputs_part4")
TABLE_DIR_PART4 = OUTPUT_DIR_PART4 / "tables"
FIGURE_DIR_PART4 = OUTPUT_DIR_PART4 / "figures"

TABLE_DIR_PART4.mkdir(parents=True, exist_ok=True)
FIGURE_DIR_PART4.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------------------------
# Validate dataset availability
if "wage1" not in globals():
    raise NameError("Run Part 1 first so that the 'wage1' DataFrame is available.")

if not isinstance(wage1, pd.DataFrame):
    raise TypeError("'wage1' must be a pandas DataFrame.")

required_variables = [
    "wage", "educ", "exper", "tenure", "female", "married", "nonwhite"
]
missing_variables = [v for v in required_variables if v not in wage1.columns]
if missing_variables:
    raise ValueError(f"Required variables missing from wage1: {missing_variables}")

# Work on a copy and preserve the original data.
diagnostic_data = wage1[required_variables].copy()
for variable in required_variables:
    diagnostic_data[variable] = pd.to_numeric(
        diagnostic_data[variable], errors="coerce"
    )
diagnostic_data = diagnostic_data.replace([np.inf, -np.inf], np.nan).dropna()

if diagnostic_data.empty:
    raise ValueError("No complete observations are available for the diagnostic model.")

if (diagnostic_data["wage"] <= 0).any():
    raise ValueError("Non-positive wages found in the diagnostic sample.")

# Use the same multiple-regression specification as Part 2.
diagnostic_model = smf.ols(
    "wage ~ educ + exper + tenure + female + married + nonwhite",
    data=diagnostic_data
).fit()

# ---------------------------------------------------------------------------
# 1. Multicollinearity: correlation matrix
numeric_predictors = diagnostic_data[
    ["educ", "exper", "tenure", "female", "married", "nonwhite"]
].copy()

correlation_matrix = numeric_predictors.corr()
display(correlation_matrix.round(4))
correlation_matrix.to_csv(TABLE_DIR_PART4 / "predictor_correlation_matrix.csv")

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5
)
plt.title("Correlation Matrix of Explanatory Variables")
plt.tight_layout()
plt.savefig(FIGURE_DIR_PART4 / "predictor_correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()

# ---------------------------------------------------------------------------
# 2. Multicollinearity: Variance Inflation Factors
X_vif = sm.add_constant(numeric_predictors, has_constant="add")
vif_table = pd.DataFrame(
    {
        "Variable": X_vif.columns,
        "VIF": [
            variance_inflation_factor(X_vif.values, i)
            for i in range(X_vif.shape[1])
        ],
    }
)
vif_table = vif_table[vif_table["Variable"] != "const"].reset_index(drop=True)
display(vif_table.round(4))
vif_table.to_csv(TABLE_DIR_PART4 / "variance_inflation_factors.csv", index=False)

# ---------------------------------------------------------------------------
# 3. Heteroskedasticity: Breusch-Pagan-Godfrey and White tests
residuals = diagnostic_model.resid
exog = diagnostic_model.model.exog

bp_lm, bp_lm_pvalue, bp_f, bp_f_pvalue = het_breuschpagan(
    residuals, exog, robust=True
)
white_lm, white_lm_pvalue, white_f, white_f_pvalue = het_white(
    residuals, exog
)

heteroskedasticity_results = pd.DataFrame(
    [
        {
            "Test": "Breusch-Pagan-Godfrey",
            "LM statistic": bp_lm,
            "LM p-value": bp_lm_pvalue,
            "F statistic": bp_f,
            "F p-value": bp_f_pvalue,
            "Reject homoskedasticity at 5%": bp_lm_pvalue < 0.05,
        },
        {
            "Test": "White",
            "LM statistic": white_lm,
            "LM p-value": white_lm_pvalue,
            "F statistic": white_f,
            "F p-value": white_f_pvalue,
            "Reject homoskedasticity at 5%": white_lm_pvalue < 0.05,
        },
    ]
)
display(heteroskedasticity_results)
heteroskedasticity_results.to_csv(
    TABLE_DIR_PART4 / "heteroskedasticity_tests.csv", index=False
)

# ---------------------------------------------------------------------------
# 4. Conventional versus heteroskedasticity-robust standard errors
conventional_model = diagnostic_model
robust_model = diagnostic_model.get_robustcov_results(cov_type="HC1")

robust_comparison = pd.DataFrame(
    {
        "Variable": diagnostic_model.params.index,
        "Coefficient": np.asarray(diagnostic_model.params),
        "Conventional standard error": np.asarray(diagnostic_model.bse),
        "Robust standard error (HC1)": np.asarray(robust_model.bse),
        "Conventional p-value": np.asarray(diagnostic_model.pvalues),
        "Robust p-value (HC1)": np.asarray(robust_model.pvalues),
    }
)
display(robust_comparison)
robust_comparison.to_csv(
    TABLE_DIR_PART4 / "conventional_vs_robust_standard_errors.csv", index=False
)

# ---------------------------------------------------------------------------
# 5. Residual normality: Jarque-Bera test
from scipy.stats import jarque_bera

jb_result = jarque_bera(residuals)
normality_results = pd.DataFrame(
    [
        {
            "Test": "Jarque-Bera",
            "Statistic": float(jb_result.statistic),
            "p-value": float(jb_result.pvalue),
            "Reject normality at 5%": float(jb_result.pvalue) < 0.05,
            "Number of residuals": len(residuals),
        }
    ]
)
display(normality_results)
normality_results.to_csv(TABLE_DIR_PART4 / "jarque_bera_normality_test.csv", index=False)

# ---------------------------------------------------------------------------
# 6. Residual Q-Q plot
sm.qqplot(residuals, line="45", fit=True)
plt.title("Normal Q-Q Plot of OLS Residuals")
plt.xlabel("Theoretical Normal Quantiles")
plt.ylabel("Ordered OLS Residuals")
plt.tight_layout()
plt.savefig(FIGURE_DIR_PART4 / "residual_normal_qq_plot.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()

# ---------------------------------------------------------------------------
# 7. Residual distribution
plt.figure(figsize=(8, 5))
sns.histplot(residuals, kde=True, stat="density")
plt.title("Distribution of OLS Residuals")
plt.xlabel("OLS Residual")
plt.ylabel("Density")
plt.tight_layout()
plt.savefig(FIGURE_DIR_PART4 / "residual_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()

# ---------------------------------------------------------------------------
# 8. Save compact model and diagnostic summary
diagnostic_summary = pd.DataFrame(
    [
        {
            "Model": "Multiple OLS: wage ~ educ + exper + tenure + female + married + nonwhite",
            "Observations": int(diagnostic_model.nobs),
            "R-squared": diagnostic_model.rsquared,
            "Adjusted R-squared": diagnostic_model.rsquared_adj,
            "Breusch-Pagan-Godfrey p-value": bp_lm_pvalue,
            "White test p-value": white_lm_pvalue,
            "Jarque-Bera p-value": jb_result.pvalue,
        }
    ]
)
display(diagnostic_summary)
diagnostic_summary.to_csv(TABLE_DIR_PART4 / "diagnostic_summary.csv", index=False)

print("Part 4 tables saved to:", TABLE_DIR_PART4.resolve())
print("Part 4 figures saved to:", FIGURE_DIR_PART4.resolve())
```

```python
# Part 5: Model Specification and Time-Series Autocorrelation

import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
from statsmodels.stats.diagnostic import linear_reset, acorr_breusch_godfrey
from statsmodels.stats.stattools import durbin_watson
from IPython.display import display
from pathlib import Path

# ---------------------------------------------------------------------------
# Output directories
OUTPUT_DIR_PART5 = Path("outputs_part5")
TABLE_DIR_PART5 = OUTPUT_DIR_PART5 / "tables"
FIGURE_DIR_PART5 = OUTPUT_DIR_PART5 / "figures"

TABLE_DIR_PART5.mkdir(parents=True, exist_ok=True)
FIGURE_DIR_PART5.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------------------------
# Model specification: Ramsey RESET and joint F-test for selected regressors
if "wage1" not in globals():
    raise NameError("Run Part 1 first so that the 'wage1' DataFrame is available.")

required_wage_variables = [
    "wage", "educ", "exper", "tenure", "female", "married", "nonwhite"
]
missing_wage_variables = [
    v for v in required_wage_variables if v not in wage1.columns
]
if missing_wage_variables:
    raise ValueError(f"Required variables missing from wage1: {missing_wage_variables}")

specification_data = wage1[required_wage_variables].copy()
for variable in required_wage_variables:
    specification_data[variable] = pd.to_numeric(
        specification_data[variable], errors="coerce"
    )
specification_data = specification_data.replace([np.inf, -np.inf], np.nan).dropna()

if specification_data.empty:
    raise ValueError("No complete observations remain for the specification model.")

specification_model = smf.ols(
    "wage ~ educ + exper + tenure + female + married + nonwhite",
    data=specification_data
).fit()

# Ramsey RESET test using fitted-value powers
reset_result = linear_reset(
    specification_model,
    power=3,
    test_type="fitted",
    use_f=True
)
reset_statistic = float(np.asarray(reset_result.fvalue).squeeze())
reset_pvalue = float(np.asarray(reset_result.pvalue).squeeze())

reset_results = pd.DataFrame(
    [{
        "Test": "Ramsey RESET",
        "Specification": "Powers of fitted values through cubic",
        "F-statistic": reset_statistic,
        "p-value": reset_pvalue,
        "Reject null at 5%": reset_pvalue < 0.05
    }]
)
display(reset_results)
reset_results.to_csv(TABLE_DIR_PART5 / "ramsey_reset_test.csv", index=False)

# Joint F-test for potentially irrelevant regressors.
# This is a pre-specified joint restriction, not a claim that variables are
# irrelevant solely because their individual p-values are large.
joint_irrelevance_test = specification_model.f_test(
    "married = 0, nonwhite = 0"
)
joint_f_stat = float(np.asarray(joint_irrelevance_test.fvalue).squeeze())
joint_f_pvalue = float(np.asarray(joint_irrelevance_test.pvalue).squeeze())
joint_df_num = int(np.asarray(joint_irrelevance_test.df_num).squeeze())
joint_df_denom = float(np.asarray(joint_irrelevance_test.df_denom).squeeze())

joint_specification_results = pd.DataFrame(
    [{
        "Test": "Joint F-test for selected regressors",
        "Restrictions": "married = 0 and nonwhite = 0",
        "F-statistic": joint_f_stat,
        "Numerator df": joint_df_num,
        "Denominator df": joint_df_denom,
        "p-value": joint_f_pvalue,
        "Reject null at 5%": joint_f_pvalue < 0.05
    }]
)
display(joint_specification_results)
joint_specification_results.to_csv(
    TABLE_DIR_PART5 / "joint_f_test_selected_regressors.csv", index=False
)

# ---------------------------------------------------------------------------
# Time-series autocorrelation exercise using phillips
if "phillips" not in globals():
    raise NameError("Run Part 1 first so that the 'phillips' DataFrame is available.")

if not isinstance(phillips, pd.DataFrame):
    raise TypeError("'phillips' must be a pandas DataFrame.")

phillips_data = phillips.copy(deep=True)
phillips_data.columns = [str(c).strip().lower() for c in phillips_data.columns]

print("Available phillips variables:")
display(pd.DataFrame({"Variable": phillips_data.columns}))

# Select documented, available numeric variables without inventing names.
numeric_columns = phillips_data.select_dtypes(include=[np.number]).columns.tolist()
if len(numeric_columns) < 2:
    raise ValueError(
        "The phillips dataset does not contain at least two numeric variables. "
        "Inspect the dataset documentation before specifying the time-series model."
    )

# Use the first two numeric columns only after displaying the actual columns.
# For the Wooldridge Phillips curve dataset, the usual variables are inflation
# and unemployment; validate column names before using them.
candidate_y = next(
    (c for c in ["inf", "infl", "inflation"] if c in phillips_data.columns),
    None
)
candidate_x = next(
    (c for c in ["unem", "unemployment", "unemp"] if c in phillips_data.columns),
    None
)

if candidate_y is None or candidate_x is None:
    raise ValueError(
        "Could not identify documented inflation and unemployment variables "
        "from the available column names. Inspect the dataset documentation "
        "and explicitly specify the model variables before continuing."
    )

time_series_data = phillips_data[[candidate_y, candidate_x]].copy()
time_series_data[candidate_y] = pd.to_numeric(
    time_series_data[candidate_y], errors="coerce"
)
time_series_data[candidate_x] = pd.to_numeric(
    time_series_data[candidate_x], errors="coerce"
)
time_series_data = time_series_data.replace([np.inf, -np.inf], np.nan).dropna()

if len(time_series_data) < 10:
    raise ValueError("Too few valid observations remain for the time-series exercise.")

# Phillips curve regression. Preserve dataset observation order; do not create
# artificial dates or reorder rows without documentation.
time_series_model = smf.ols(
    f'Q("{candidate_y}") ~ Q("{candidate_x}")',
    data=time_series_data
).fit()

# Report the data ordering assumption explicitly.
time_index_information = pd.DataFrame(
    [{
        "Observations used": len(time_series_data),
        "Dependent variable": candidate_y,
        "Explanatory variable": candidate_x,
        "Ordering": "Original dataset row order",
        "Time index verified from metadata": False,
        "Interpretation note": (
            "Autocorrelation tests require meaningful temporal ordering. "
            "Confirm the dataset's documented time index before interpreting "
            "these results as time-series evidence."
        )
    }]
)
display(time_index_information)
time_index_information.to_csv(
    TABLE_DIR_PART5 / "phillips_time_series_data_notes.csv", index=False
)

# Plot the variables in source row order; this is only a row-order plot until
# the documented time index is verified.
fig, axes = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
axes[0].plot(time_series_data[candidate_y].to_numpy())
axes[0].set_title(f"{candidate_y} by observation order")
axes[0].set_ylabel(candidate_y)
axes[1].plot(time_series_data[candidate_x].to_numpy())
axes[1].set_title(f"{candidate_x} by observation order")
axes[1].set_ylabel(candidate_x)
axes[1].set_xlabel("Observation order (not a verified calendar time index)")
plt.tight_layout()
plt.savefig(FIGURE_DIR_PART5 / "phillips_variables_by_observation_order.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()

# Time-series model output
display(time_series_model.summary())

# Durbin-Watson statistic
dw_statistic = float(durbin_watson(time_series_model.resid))

# Breusch-Godfrey LM test for up to 4 lags, if sample size permits.
bg_lags = min(4, max(1, len(time_series_data) // 5))
bg_result = acorr_breusch_godfrey(
    time_series_model,
    nlags=bg_lags,
    store=False
)
bg_lm_stat, bg_lm_pvalue, bg_f_stat, bg_f_pvalue = bg_result

autocorrelation_results = pd.DataFrame(
    [{
        "Test": "Durbin-Watson",
        "Statistic": dw_statistic,
        "p-value": np.nan,
        "Lag order": 1,
        "Interpretation": (
            "The Durbin-Watson statistic is descriptive; use its appropriate "
            "critical-value bounds or a suitable formal procedure for inference."
        )
    }, {
        "Test": "Breusch-Godfrey",
        "Statistic": float(bg_lm_stat),
        "p-value": float(bg_lm_pvalue),
        "Lag order": bg_lags,
        "Interpretation": (
            "Tests the null of no residual autocorrelation up to the specified lag order."
        )
    }]
)
display(autocorrelation_results)
autocorrelation_results.to_csv(
    TABLE_DIR_PART5 / "autocorrelation_tests.csv", index=False
)

bg_f_results = pd.DataFrame(
    [{
        "Test": "Breusch-Godfrey F form",
        "F-statistic": float(bg_f_stat),
        "p-value": float(bg_f_pvalue),
        "Lag order": bg_lags
    }]
)
display(bg_f_results)
bg_f_results.to_csv(TABLE_DIR_PART5 / "breusch_godfrey_f_test.csv", index=False)

# Save model summaries
with open(OUTPUT_DIR_PART5 / "wage_specification_model_summary.txt", "w", encoding="utf-8") as f:
    f.write(specification_model.summary().as_text())

with open(OUTPUT_DIR_PART5 / "phillips_time_series_model_summary.txt", "w", encoding="utf-8") as f:
    f.write(time_series_model.summary().as_text())

print("Part 5 tables and model summaries saved to:", OUTPUT_DIR_PART5.resolve())
```

```python
"""Part 6: Consolidate all project results into a single Excel workbook and final report.

Run after Parts 1–5. The script collects available CSV tables and figures from
outputs_part1 through outputs_part5, creates a multi-sheet Excel workbook, writes
a Markdown report and README, and packages the final deliverables in a ZIP file.
"""

from pathlib import Path
from datetime import date
import json
import re
import shutil
import zipfile
import pandas as pd

PROJECT_DIR = Path.cwd()
FINAL_DIR = PROJECT_DIR / "final_project_outputs"
TABLE_DIR = FINAL_DIR / "tables"
FIGURE_DIR = FINAL_DIR / "figures"
REPORT_DIR = FINAL_DIR / "report"
for directory in (FINAL_DIR, TABLE_DIR, FIGURE_DIR, REPORT_DIR):
    directory.mkdir(parents=True, exist_ok=True)

PART_FOLDERS = {
    "Part 1 — Data and EDA": PROJECT_DIR / "outputs_part1",
    "Part 2 — OLS and inference": PROJECT_DIR / "outputs_part2",
    "Part 3 — Functional forms and dummy variables": PROJECT_DIR / "outputs_part3",
    "Part 4 — Regression diagnostics": PROJECT_DIR / "outputs_part4",
    "Part 5 — Specification and autocorrelation": PROJECT_DIR / "outputs_part5",
}

def safe_name(value, limit=31):
    value = re.sub(r"[^A-Za-z0-9_]+", "_", str(value)).strip("_")
    return (value or "Results")[:limit]

def markdown_table(frame, max_rows=30):
    if frame is None or frame.empty:
        return "*No corresponding output table was found.*"
    view = frame.head(max_rows).copy()
    view = view.loc[:, list(view.columns[:10])]
    try:
        return view.to_markdown(index=False)
    except Exception:
        return "\n\n" + view.to_string(index=False) + "\n\n"

collected = []
loaded_tables = {}
copied_figures = []
unreadable_files = []

# Collect every CSV table and figure from each part's output folder.
for part_name, folder in PART_FOLDERS.items():
    if not folder.exists():
        print(f"Output folder not found: {folder} (will be noted in the report)")
        continue

    for csv_path in sorted(folder.rglob("*.csv")):
        try:
            frame = pd.read_csv(csv_path)
            relative = f"{folder.name}/{csv_path.relative_to(folder).as_posix()}"
            safe_stem = safe_name(f"{folder.name}_{csv_path.stem}", 100)
            destination = TABLE_DIR / f"{safe_stem}.csv"
            frame.to_csv(destination, index=False)
            loaded_tables[relative] = frame
            collected.append({
                "part": part_name,
                "source_file": relative,
                "rows": len(frame),
                "columns": len(frame.columns),
                "consolidated_file": str(destination.relative_to(PROJECT_DIR)),
                "status": "collected",
            })
        except Exception as exc:
            unreadable_files.append(f"{csv_path}: {exc}")
            collected.append({
                "part": part_name, "source_file": str(csv_path),
                "rows": None, "columns": None, "consolidated_file": "",
                "status": f"Could not read CSV: {exc}",
            })

    for image_path in sorted(folder.rglob("*")):
        if image_path.is_file() and image_path.suffix.lower() in {
            ".png", ".jpg", ".jpeg", ".svg", ".webp"
        }:
            try:
                rel = image_path.relative_to(folder)
                destination = FIGURE_DIR / safe_name(folder.name, 30) / rel
                destination.parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(image_path, destination)
                copied_figures.append(str(destination.relative_to(PROJECT_DIR)))
            except Exception as exc:
                unreadable_files.append(f"{image_path}: {exc}")

inventory = pd.DataFrame(collected)
inventory_path = FINAL_DIR / "consolidated_table_inventory.csv"
inventory.to_csv(inventory_path, index=False)

# A single Excel workbook with one worksheet per collected CSV, plus an index.
workbook_path = FINAL_DIR / "All_Project_Results.xlsx"
with pd.ExcelWriter(workbook_path, engine="openpyxl") as writer:
    index_rows = []
    used_sheets = set()
    for source_name, frame in loaded_tables.items():
        sheet_base = safe_name(Path(source_name).stem)
        sheet = sheet_base
        suffix = 2
        while sheet in used_sheets:
            tail = f"_{suffix}"
            sheet = sheet_base[:31-len(tail)] + tail
            suffix += 1
        used_sheets.add(sheet)
        frame.to_excel(writer, sheet_name=sheet, index=False)
        index_rows.append({
            "worksheet": sheet,
            "source_file": source_name,
            "rows": len(frame),
            "columns": len(frame.columns),
        })
    pd.DataFrame(index_rows, columns=[
        "worksheet", "source_file", "rows", "columns"
    ]).to_excel(writer, sheet_name="Results_Index", index=False)
    inventory.to_excel(writer, sheet_name="File_Inventory", index=False)

def find_table(name_fragment):
    fragment = name_fragment.lower()
    for source_name, frame in loaded_tables.items():
        if fragment in Path(source_name).name.lower():
            return source_name, frame
    return None, None

def table_section(title, filename_fragment, explanation):
    source_name, frame = find_table(filename_fragment)
    lines = [f"### {title}", "", explanation, ""]
    if source_name:
        lines.extend([f"Source output: `{source_name}`", "", markdown_table(frame), ""])
    else:
        lines.extend(["No matching CSV was found in the collected outputs.", ""])
    return "\n".join(lines)

part_status = []
for part_name, folder in PART_FOLDERS.items():
    files_found = len(list(folder.rglob("*"))) if folder.exists() else 0
    part_status.append(
        f"- **{part_name}:** {'output folder found' if folder.exists() else 'output folder not found'} "
        f"({files_found} filesystem entries)"
    )

report = f"""# Applied Econometrics Project: Wage Determination

**Generated:** {date.today().isoformat()}
**Primary dataset:** Wooldridge `wage1` (cross-sectional wage data)
**Supplementary dataset:** Wooldridge `phillips` (time-series inflation and unemployment analysis)

## 1. Research objective

This project applies econometric methods to wage determination, compares alternative specifications, evaluates regression diagnostics, and examines autocorrelation in a supplementary time-series model.

## 2. Coverage of Parts 1–5

{chr(10).join(part_status)}

## 3. Data preparation and exploratory analysis — Part 1

{table_section('Descriptive statistics', 'descriptive', 'Descriptive statistics summarise the distributions of the variables used in the project.')}

{table_section('Missing values and data quality', 'missing', 'Review missingness and data-quality checks before interpreting estimates.')}

Part 1 figures, along with figures from Parts 2–5, are copied to `final_project_outputs/figures/` when present.

## 4. OLS estimation and inference — Part 2

{table_section('OLS model fit comparison', 'ols_model_fit_comparison', 'Compare sample size and model-level fit statistics across wage regressions.')}

{table_section('Individual coefficient tests', 'individual_t_tests', 'Use estimates, standard errors, test statistics, and p-values to assess individual hypotheses.')}

{table_section('Joint test for experience and tenure', 'joint_f_test_experience_tenure', 'The joint test evaluates whether the selected coefficients are jointly zero under the reported model.')}

## 5. Functional forms and indicator variables — Part 3

{table_section('Functional-form comparison', 'functional_form_model', 'Compare model specifications while accounting for changes in the dependent-variable scale.')}

{table_section('Dummy-variable and interaction results', 'interaction', 'Interpret indicator variables relative to their reference categories and interactions relative to the model baseline.')}

## 6. Regression diagnostics — Part 4

{table_section('Heteroskedasticity tests', 'heteroskedasticity', 'Breusch–Pagan and White tests assess whether residual variance may be non-constant.')}

{table_section('Conventional and robust standard errors', 'robust_standard', 'Compare conventional and heteroskedasticity-consistent standard errors; robust inference does not change OLS coefficients.')}

{table_section('Variance inflation factors', 'variance_inflation_factors', 'VIF values indicate how strongly each regressor is linearly explained by other regressors.')}

## 7. Model specification and autocorrelation — Part 5

{table_section('Ramsey RESET specification test', 'ramsey_reset_test', 'RESET may indicate functional-form or omitted-variable problems but does not identify a unique cause.')}

{table_section('Autocorrelation tests', 'autocorrelation', 'Interpret Durbin–Watson and Breusch–Godfrey results in light of model specification, time ordering, and lag order.')}

## 8. Consolidated results file

The file `All_Project_Results.xlsx` contains every readable CSV table found in the output folders for Parts 1–5. Each table is placed on a separate worksheet, with `Results_Index` and `File_Inventory` sheets for navigation. Original CSV copies are stored in `tables/`.

- CSV tables collected: **{len(loaded_tables)}**
- Figures copied: **{len(copied_figures)}**
- CSV or figure processing issues: **{len(unreadable_files)}**

## 9. Interpretation checklist

- Report coefficient units and the dependent-variable scale for each specification.
- For log-dependent-variable models, distinguish approximate percentage effects from exact percentage effects where appropriate.
- Report p-values alongside effect sizes and confidence intervals; statistical significance alone does not establish economic importance.
- If heteroskedasticity is detected, consider robust standard errors for inference.
- Assess multicollinearity using correlation patterns, VIF values, sample size, and substantive model structure together.
- Treat RESET as a general diagnostic rather than proof of a particular omitted variable.
- Confirm chronological ordering and the time index of `phillips` before making time-series claims.
- Avoid causal claims unless the research design and identifying assumptions justify them.

## 10. Missing outputs

Only files present in the output folders at the time of execution are included. If a part's output folder is missing or a desired table is absent, run that part and rerun Part 6. The script does not invent missing estimates or test results.

## 11. Reproducibility

See `README.md` and `requirements.txt` in the project root. Figures and consolidated CSV tables are included in the final output directory.
"""

report_path = REPORT_DIR / "final_econometrics_report.md"
report_path.write_text(report, encoding="utf-8")
# Also provide a plain-text copy for easy viewing in any editor.
(REPORT_DIR / "final_econometrics_report.txt").write_text(report, encoding="utf-8")

readme = """# Applied Econometrics: Wage Determination

A reproducible applied econometrics project using Wooldridge's `wage1` cross-sectional dataset and `phillips` time-series dataset.

## Project structure

- `Part1_Wage_Determination_EDA_clean.py` — data loading, descriptive analysis, and visualisation.
- `Part2_OLS_and_Inference.py` — OLS estimation and hypothesis testing.
- `Part3_Functional_Forms_and_Dummy_Variables.py` — functional forms, indicators, and interactions.
- `Part4_Classical_Regression_Diagnostics.py` — multicollinearity, heteroskedasticity, and residual diagnostics.
- `Part5_Model_Specification_and_Autocorrelation.py` — specification and autocorrelation tests.
- `Part6_Final_Results_Report_and_GitHub.py` — consolidates results and creates the final deliverables.
- `final_project_outputs/All_Project_Results.xlsx` — all available CSV result tables in one Excel workbook.
- `final_project_outputs/report/` — final Markdown and plain-text reports.
- `final_project_outputs/tables/` and `figures/` — consolidated CSV tables and figures.

## Run in Google Colab

1. Run Parts 1–5 in order in the same working directory.
2. Run Part 6.
3. Download `final_project_outputs/All_Project_Results.xlsx` for all tabular results in one file.
4. Download `final_project_outputs.zip` for the full package of reports, tables, and figures.

Only outputs actually produced by Parts 1–5 are collected. Run a part again if its output folder or expected tables are missing.
"""
(PROJECT_DIR / "README.md").write_text(readme, encoding="utf-8")
(PROJECT_DIR / "requirements.txt").write_text(
    "numpy\npandas\nstatsmodels\nscipy\nmatplotlib\nseaborn\nwooldridge\nopenpyxl\ntabulate\n",
    encoding="utf-8",
)

manifest = {
    "report_generated": date.today().isoformat(),
    "project_directory": str(PROJECT_DIR),
    "source_tables_collected": len(loaded_tables),
    "figures_copied": len(copied_figures),
    "excel_workbook": str(workbook_path),
    "report_markdown": str(report_path),
    "inventory": str(inventory_path),
    "tables_directory": str(TABLE_DIR),
    "figures_directory": str(FIGURE_DIR),
    "processing_issues": unreadable_files,
}
(FINAL_DIR / "generation_manifest.json").write_text(json.dumps(manifest, indent=2), encoding="utf-8")

# Create a single ZIP package containing the workbook, report, CSV tables, figures,
# manifest, README, and requirements. Exclude the ZIP itself if rerun.
zip_path = PROJECT_DIR / "final_project_outputs.zip"
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as archive:
    for path in FINAL_DIR.rglob("*"):
        if path.is_file() and path != zip_path:
            archive.write(path, arcname=Path("final_project_outputs") / path.relative_to(FINAL_DIR))
    for name in ("README.md", "requirements.txt"):
        path = PROJECT_DIR / name
        if path.exists():
            archive.write(path, arcname=name)

print("Part 6 completed.")
print(f"Result tables collected: {len(loaded_tables)}")
print(f"Figures copied: {len(copied_figures)}")
print(f"Excel workbook with all available result tables: {workbook_path}")
print(f"Final report: {report_path}")
print(f"Full project package: {zip_path}")
if unreadable_files:
    print(f"Files that could not be processed: {len(unreadable_files)} (see generation_manifest.json)")
```

```python
# Comprehensive Econometrics Project Report Generator
# Creates an editable Word report and a PDF containing:
# - narrative and data-driven interpretation
# - every available exported results table
# - every available figure
# - full source code for Parts 1-6 (when the .py files are present)
# - a contents/inventory section identifying missing artifacts

from pathlib import Path
import re
import sys
import subprocess
import textwrap
import pandas as pd
import numpy as np

# Install required document-generation packages if needed.
for module, package in [
    ("docx", "python-docx"),
    ("reportlab", "reportlab"),
    ("PIL", "pillow"),
]:
    try:
        __import__(module)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])

from docx import Document
from docx.shared import Inches, Pt, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_CELL_VERTICAL_ALIGNMENT
from docx.oxml import OxmlElement
from docx.oxml.ns import qn
from PIL import Image as PILImage

from reportlab.lib import colors
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.pagesizes import A4, landscape
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.units import cm
from reportlab.platypus import (
    SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
    Image as RLImage, PageBreak, KeepTogether, Preformatted
)
from xml.sax.saxutils import escape

# ---------------------------------------------------------------------------
# Paths and configuration
# ---------------------------------------------------------------------------
CONTENT = Path("/content")
OUTPUT_ROOT = CONTENT / "final_project_outputs"
REPORT_DIR = OUTPUT_ROOT / "report"
TABLE_DIR = OUTPUT_ROOT / "tables"
FIGURE_DIRS = [
    OUTPUT_ROOT / "figures",
    OUTPUT_ROOT / "figures_part1",
    CONTENT / "figures_part1",
    CONTENT / "figures",
]
CODE_DIRS = [CONTENT, CONTENT / "code", CONTENT / "project_code"]

REPORT_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)

DOCX_PATH = REPORT_DIR / "comprehensive_econometrics_project_report.docx"
PDF_PATH = REPORT_DIR / "comprehensive_econometrics_project_report.pdf"
SUMMARY_PATH = REPORT_DIR / "report_generation_inventory.csv"

# ---------------------------------------------------------------------------
# Find result tables and figures
# ---------------------------------------------------------------------------
csv_files = []
for base in [OUTPUT_ROOT, CONTENT / "outputs_part1", CONTENT / "outputs_part2",
             CONTENT / "outputs_part3", CONTENT / "outputs_part4",
             CONTENT / "outputs_part5"]:
    if base.exists():
        csv_files.extend(p for p in base.rglob("*.csv") if p.is_file())

# Deduplicate by resolved path, while excluding any previous report inventory.
unique_csv = {}
for p in csv_files:
    if p.name != SUMMARY_PATH.name:
        unique_csv[str(p.resolve())] = p
csv_files = sorted(unique_csv.values(), key=lambda p: str(p).lower())

image_files = []
seen_images = set()
for base in FIGURE_DIRS + [OUTPUT_ROOT]:
    if base.exists():
        for p in sorted(base.rglob("*")):
            if p.is_file() and p.suffix.lower() in {".png", ".jpg", ".jpeg", ".webp"}:
                rp = str(p.resolve())
                if rp not in seen_images:
                    seen_images.add(rp)
                    image_files.append(p)

# ---------------------------------------------------------------------------
# Locate source code files for the six parts
# ---------------------------------------------------------------------------
expected_code_names = {
    1: [
        "Part1_Wage_Determination_EDA_clean.py",
        "Part1_Wage_Determination_EDA_code_only_fixed.py",
        "Part1_Wage_Determination_EDA_code_only.py",
    ],
    2: ["Part2_OLS_and_Inference.py"],
    3: [
        "Part3_Functional_Forms_and_Dummy_Variables.py",
        "Part3_Functional_Forms_and_Dummies.py",
    ],
    4: ["Part4_Classical_Regression_Diagnostics.py"],
    5: ["Part5_Model_Specification_and_Autocorrelation.py"],
    6: [
        "Part6_Final_Results_Report_and_GitHub_revised.py",
        "Part6_Final_Results_Report_and_GitHub.py",
    ],
}

all_py = []
for base in CODE_DIRS:
    if base.exists():
        all_py.extend(base.rglob("*.py"))

def find_code(part_num):
    candidates = expected_code_names[part_num]
    # Prefer the first existing filename in the ordered candidate list.
    for name in candidates:
        matches = [p for p in all_py if p.name == name]
        if matches:
            # Prefer revised/fixed/clean variants by candidate order.
            return sorted(matches, key=lambda p: len(str(p)))[0]
    # Fallback to a loose filename match.
    part_prefix = f"Part{part_num}"
    matches = [p for p in all_py if p.name.startswith(part_prefix)]
    if matches:
        return sorted(matches, key=lambda p: ("revised" not in p.name.lower(),
                                               "fixed" not in p.name.lower(),
                                               "clean" not in p.name.lower(),
                                               len(p.name)))[0]
    return None

code_files = {n: find_code(n) for n in range(1, 7)}

# ---------------------------------------------------------------------------
# Helpers: labels, table interpretation, and formatting
# ---------------------------------------------------------------------------
def nice_name(path_or_name):
    name = Path(path_or_name).stem
    name = re.sub(r"^part\d+[_-]*", "", name, flags=re.I)
    name = name.replace("_", " ").replace("-", " ")
    return re.sub(r"\s+", " ", name).strip().title()

def fmt_value(x):
    if pd.isna(x):
        return ""
    if isinstance(x, (float, np.floating)):
        if abs(x) >= 10000 or (0 < abs(x) < 0.001):
            return f"{x:.3e}"
        return f"{x:.4f}".rstrip("0").rstrip(".")
    return str(x)

def read_csv_safely(path):
    try:
        return pd.read_csv(path)
    except Exception:
        try:
            return pd.read_csv(path, index_col=0)
        except Exception as exc:
            return None

def data_interpretation(name, df):
    """Produce cautious, table-specific descriptions from the actual table values."""
    lines = []
    if df is None or df.empty:
        return ["The file is present but could not be read as a non-empty CSV table."]
    lower_cols = {str(c).lower(): c for c in df.columns}
    nrows, ncols = df.shape
    lines.append(f"This table contains {nrows} rows and {ncols} columns.")

    # Dataset summary tables
    if "observations" in lower_cols and "variables" in lower_cols:
        obs_col, var_col = lower_cols["observations"], lower_cols["variables"]
        for _, row in df.iterrows():
            label = row.iloc[0]
            lines.append(
                f"{label}: {fmt_value(row[obs_col])} observations and "
                f"{fmt_value(row[var_col])} variables."
            )
        lines.append("These counts describe the datasets used for the project's cross-sectional wage analysis and supplementary time-series analysis.")
        return lines

    # Coefficient/regression tables
    coeff_names = [c for c in df.columns if str(c).lower() in
                   {"coef", "coefficient", "estimate", "std err", "std_error", "p>|t|", "pvalue", "p_value"}]
    if coeff_names:
        p_col = next((lower_cols[k] for k in ["p>|t|", "pvalue", "p_value", "p-value"] if k in lower_cols), None)
        coef_col = next((lower_cols[k] for k in ["coef", "coefficient", "estimate"] if k in lower_cols), None)
        term_col = next((c for c in df.columns if str(c).lower() in {"term", "variable", "parameter", "feature"}), df.columns[0])
        if coef_col:
            for _, row in df.iterrows():
                term = str(row.get(term_col, ""))
                try:
                    coef = float(row[coef_col])
                except Exception:
                    continue
                msg = f"Estimated coefficient for {term} is {coef:.4f}"
                if p_col:
                    try:
                        pv = float(row[p_col])
                        msg += f" (p-value {pv:.4g})"
                        if pv < 0.05:
                            msg += "; statistically significant at the 5% level under the reported model"
                        else:
                            msg += "; not statistically significant at the 5% level under the reported model"
                    except Exception:
                        pass
                lines.append(msg + ".")
            return lines

    # P-value / test tables
    p_cols = [c for c in df.columns if any(s in str(c).lower() for s in ["pvalue", "p-value", "p_value", "p>|", "prob"])]
    if p_cols:
        for _, row in df.iterrows():
            for pc in p_cols:
                try:
                    p = float(row[pc])
                    label = str(row.iloc[0])
                    conclusion = ("reject the null hypothesis at the 5% level"
                                  if p < 0.05 else
                                  "do not reject the null hypothesis at the 5% level")
                    lines.append(f"{label}: {pc} = {p:.4g}; {conclusion}, conditional on the test's assumptions.")
                except Exception:
                    pass
        if len(lines) > 1:
            return lines

    # Generic numeric tables
    numeric = df.select_dtypes(include=[np.number])
    if not numeric.empty:
        for col in numeric.columns[:8]:
            series = numeric[col].dropna()
            if not series.empty:
                lines.append(
                    f"{col}: {len(series)} non-missing numeric entries; "
                    f"minimum {series.min():.4g}, maximum {series.max():.4g}, "
                    f"mean {series.mean():.4g}."
                )
    else:
        lines.append("This is a categorical/text summary table; interpret its entries using the row and column labels.")
    lines.append(
        "Interpretation is descriptive of the exported values; statistical or causal claims should be based on the relevant model specification and assumptions."
    )
    return lines

# Inventory before writing the inventory itself
inventory_rows = []
for p in csv_files:
    df = read_csv_safely(p)
    inventory_rows.append({
        "artifact_type": "CSV result table",
        "name": p.name,
        "path": str(p),
        "readable": df is not None,
        "rows": None if df is None else len(df),
        "columns": None if df is None else len(df.columns),
    })
for p in image_files:
    inventory_rows.append({
        "artifact_type": "Figure",
        "name": p.name,
        "path": str(p),
        "readable": True,
        "rows": "",
        "columns": "",
    })
for n, p in code_files.items():
    inventory_rows.append({
        "artifact_type": f"Part {n} source code",
        "name": p.name if p else f"Part{n} source not found",
        "path": str(p) if p else "",
        "readable": bool(p),
        "rows": "",
        "columns": "",
    })
pd.DataFrame(inventory_rows).to_csv(SUMMARY_PATH, index=False)

# ---------------------------------------------------------------------------
# Build Word document
# ---------------------------------------------------------------------------
doc = Document()
sec = doc.sections[0]
sec.top_margin = Inches(0.65)
sec.bottom_margin = Inches(0.65)
sec.left_margin = Inches(0.7)
sec.right_margin = Inches(0.7)

styles = doc.styles
styles["Normal"].font.name = "Arial"
styles["Normal"].font.size = Pt(9)
for nm, size in [("Title", 24), ("Heading 1", 17), ("Heading 2", 13), ("Heading 3", 10)]:
    styles[nm].font.name = "Arial"
    styles[nm].font.size = Pt(size)

# Cover page
p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
r = p.add_run("APPLIED ECONOMETRICS PROJECT")
r.bold = True
r.font.size = Pt(23)
r.font.color.rgb = RGBColor(35, 75, 115)

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
r = p.add_run("Wage Determination, Regression Diagnostics and Time-Series Autocorrelation")
r.bold = True
r.font.size = Pt(15)

p = doc.add_paragraph()
p.alignment = WD_ALIGN_PARAGRAPH.CENTER
p.add_run("Comprehensive report with results, tables, figures, interpretations and source code")

doc.add_paragraph()
doc.add_heading("Report scope", level=1)
doc.add_paragraph(
    "This report assembles the available outputs generated by the six-part project. "
    "It includes exported tables, figures located in the project folders, concise data-driven "
    "descriptions of each table, and the source code files found in the Colab working directory. "
    "The primary dataset is Wooldridge's wage1 cross-section; phillips is used for the supplementary "
    "time-series/autocorrelation exercise. Interpretations are limited to the exported results and "
    "do not treat descriptive associations as causal effects."
)
doc.add_heading("Project structure", level=1)
for title, description in [
    ("Part 1 — Data preparation and exploratory analysis",
     "Dataset verification, data quality, descriptive statistics, distributions, group comparisons and correlations."),
    ("Part 2 — OLS and inference",
     "Simple and multiple regressions, coefficient inference, hypothesis tests and fit measures."),
    ("Part 3 — Functional forms and dummy variables",
     "Alternative functional forms, indicator variables and interaction specifications."),
    ("Part 4 — Classical regression diagnostics",
     "Multicollinearity, heteroskedasticity, robust standard errors and residual diagnostics."),
    ("Part 5 — Model specification and autocorrelation",
     "Specification testing and serial-correlation tests using the supplementary time-series dataset."),
    ("Part 6 — Consolidation and documentation",
     "Compilation of results, project report and repository documentation."),
]:
    doc.add_heading(title, level=2)
    doc.add_paragraph(description)

doc.add_heading("Artifact inventory", level=1)
doc.add_paragraph(
    f"Found {len(csv_files)} CSV tables, {len(image_files)} figures, and "
    f"{sum(p is not None for p in code_files.values())} of 6 source-code files. "
    "Missing source files are explicitly noted in the code appendix."
)

doc.add_heading("Results tables and interpretations", level=1)
if not csv_files:
    doc.add_paragraph(
        "No CSV result tables were found. Run Parts 1–5 and Part 6 in the same Colab runtime, "
        "and check the output folders before generating this report."
    )

for idx, path in enumerate(csv_files, start=1):
    df = read_csv_safely(path)
    doc.add_heading(f"Table {idx}. {nice_name(path.name)}", level=2)
    doc.add_paragraph(f"Source file: {path}")
    if df is None:
        doc.add_paragraph("The CSV could not be parsed. The source file is retained in the project package.")
        continue
    if df.empty:
        doc.add_paragraph("This exported table is empty.")
        continue

    # Keep the table readable in Word; for very wide tables show all rows/cols but shrink font.
    show_df = df.copy().fillna("")
    max_rows = 100
    truncated = len(show_df) > max_rows
    if truncated:
        show_df = show_df.head(max_rows)
    word_table = doc.add_table(rows=1, cols=max(1, len(show_df.columns)))
    word_table.alignment = WD_TABLE_ALIGNMENT.CENTER
    word_table.style = "Light Shading Accent 1"
    for j, col in enumerate(show_df.columns):
        word_table.rows[0].cells[j].text = str(col)
    for _, row in show_df.iterrows():
        cells = word_table.add_row().cells
        for j, value in enumerate(row.tolist()):
            cells[j].text = fmt_value(value)
            cells[j].vertical_alignment = WD_CELL_VERTICAL_ALIGNMENT.CENTER
    for row in word_table.rows:
        for cell in row.cells:
            for para in cell.paragraphs:
                for run in para.runs:
                    run.font.size = Pt(7)
    if truncated:
        doc.add_paragraph(f"Display truncated to the first {max_rows} rows; the complete table remains in the accompanying CSV/workbook.")
    doc.add_heading("Interpretation", level=3)
    for statement in data_interpretation(path.stem, df):
        doc.add_paragraph(statement, style="Normal")

doc.add_page_break()
doc.add_heading("Figures", level=1)
if not image_files:
    doc.add_paragraph(
        "No figures were found in the configured project folders. Check /content/figures_part1 "
        "and /content/final_project_outputs/figures."
    )
for idx, path in enumerate(image_files, start=1):
    doc.add_heading(f"Figure {idx}. {nice_name(path.name)}", level=2)
    doc.add_paragraph(f"Source image: {path}")
    try:
        with PILImage.open(path) as im:
            width, height = im.size
        max_w, max_h = 6.2, 5.7
        ratio = min(max_w / (width / 96), max_h / (height / 96))
        doc.add_picture(str(path), width=Inches((width / 96) * ratio),
                        height=Inches((height / 96) * ratio))
    except Exception as exc:
        doc.add_paragraph(f"Could not insert image: {exc}")
    doc.add_paragraph(
        "Interpretation: read this figure together with its title, axis labels, and the corresponding "
        "descriptive or regression results. The figure is evidence of the displayed pattern, not by itself "
        "proof of a causal relationship."
    )

doc.add_page_break()
doc.add_heading("Source code appendix", level=1)
doc.add_paragraph(
    "The code appendix reproduces the source files found in the Colab working directory. "
    "For reproducibility, retain the original scripts alongside the data-output archive. "
    "If a part's source file was not found, it is listed as missing rather than reconstructed."
)

for part_num in range(1, 7):
    source = code_files[part_num]
    doc.add_heading(f"Part {part_num} source code", level=2)
    if source is None:
        doc.add_paragraph(
            f"Source file for Part {part_num} was not found in /content, /content/code, "
            "or /content/project_code. Upload/copy that .py file into one of those folders "
            "and rerun this report generator to include it."
        )
        continue
    doc.add_paragraph(f"File: {source.name}")
    source_text = source.read_text(encoding="utf-8", errors="replace")
    for line in source_text.splitlines():
        p = doc.add_paragraph()
        p.paragraph_format.space_after = Pt(0)
        p.paragraph_format.left_indent = Inches(0.12)
        run = p.add_run(line if line else " ")
        run.font.name = "Courier New"
        run.font.size = Pt(6.5)

doc.add_heading("Reproducibility and limitations", level=1)
doc.add_paragraph(
    "This report consolidates files that exist at the time of generation. A table or figure that was "
    "displayed in a notebook but never exported to disk cannot be included automatically. Some "
    "interpretations are generated from table labels and numerical values; they should be reviewed "
    "against the corresponding model specification before academic submission. Statistical significance "
    "does not establish causality, and diagnostic tests should be interpreted in light of their assumptions."
)
doc.save(DOCX_PATH)

# ---------------------------------------------------------------------------
# Build PDF report from the same tables, figures, and code
# ---------------------------------------------------------------------------
pdf_styles = getSampleStyleSheet()
pdf_styles.add(ParagraphStyle(
    name="CoverTitleX", parent=pdf_styles["Title"], fontSize=22, leading=27,
    alignment=TA_CENTER, spaceAfter=16
))
pdf_styles.add(ParagraphStyle(
    name="H1X", parent=pdf_styles["Heading1"], fontSize=15, leading=19,
    spaceBefore=12, spaceAfter=7
))
pdf_styles.add(ParagraphStyle(
    name="H2X", parent=pdf_styles["Heading2"], fontSize=10.5, leading=13,
    spaceBefore=8, spaceAfter=4
))
pdf_styles.add(ParagraphStyle(
    name="BodyX", parent=pdf_styles["BodyText"], fontSize=8, leading=10.5,
    spaceAfter=5
))
pdf_styles.add(ParagraphStyle(
    name="CodeX", fontName="Courier", fontSize=5.3, leading=6.4,
    leftIndent=4, rightIndent=4, spaceAfter=0
))
pdf_styles.add(ParagraphStyle(
    name="CaptionX", parent=pdf_styles["BodyText"], fontSize=7.5,
    leading=9, alignment=TA_CENTER, spaceAfter=8
))

story = []
story.append(Spacer(1, 2.2 * cm))
story.append(Paragraph("APPLIED ECONOMETRICS PROJECT", pdf_styles["CoverTitleX"]))
story.append(Paragraph(
    "Wage Determination, Regression Diagnostics and Time-Series Autocorrelation",
    pdf_styles["H1X"]
))
story.append(Spacer(1, 0.5 * cm))
story.append(Paragraph(
    "Comprehensive report with available results, tables, figures, interpretations and source code",
    pdf_styles["BodyX"]
))
story.append(Spacer(1, 0.8 * cm))
story.append(Paragraph("Report scope", pdf_styles["H1X"]))
story.append(Paragraph(
    "This report assembles available outputs from the six-part project. It includes exported tables, "
    "figures found in the project folders, data-driven descriptions, and source code files available "
    "in the Colab working directory. Interpretations are limited to the exported values and do not treat "
    "descriptive associations as causal effects.", pdf_styles["BodyX"]
))
story.append(Paragraph("Contents", pdf_styles["H1X"]))
for title in [
    "Part 1 — Data preparation and exploratory analysis",
    "Part 2 — OLS and inference",
    "Part 3 — Functional forms and dummy variables",
    "Part 4 — Classical regression diagnostics",
    "Part 5 — Model specification and autocorrelation",
    "Part 6 — Consolidation and documentation",
    "Results tables and interpretations",
    "Figures",
    "Source code appendix",
]:
    story.append(Paragraph("• " + escape(title), pdf_styles["BodyX"]))
story.append(PageBreak())

def pdf_table_for_df(df, max_rows=100, max_cols=8):
    view = df.copy().fillna("")
    was_truncated = len(view) > max_rows or len(view.columns) > max_cols
    view = view.iloc[:max_rows, :max_cols]
    data = [[Paragraph(escape(str(c)), pdf_styles["BodyX"]) for c in view.columns]]
    for _, row in view.iterrows():
        data.append([
            Paragraph(escape(fmt_value(v)), pdf_styles["BodyX"])
            for v in row.tolist()
        ])
    ncols = max(1, len(view.columns))
    page_width = A4[0] - 3.0 * cm
    table = Table(data, colWidths=[page_width / ncols] * ncols, repeatRows=1)
    table.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#D9E5F2")),
        ("GRID", (0, 0), (-1, -1), 0.35, colors.grey),
        ("VALIGN", (0, 0), (-1, -1), "TOP"),
        ("LEFTPADDING", (0, 0), (-1, -1), 3),
        ("RIGHTPADDING", (0, 0), (-1, -1), 3),
        ("TOPPADDING", (0, 0), (-1, -1), 3),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 3),
    ]))
    return table, was_truncated

story.append(Paragraph("Results tables and interpretations", pdf_styles["H1X"]))
for idx, path in enumerate(csv_files, start=1):
    df = read_csv_safely(path)
    story.append(Paragraph(f"Table {idx}. {escape(nice_name(path.name))}", pdf_styles["H2X"]))
    story.append(Paragraph(f"Source: {escape(str(path))}", pdf_styles["BodyX"]))
    if df is None or df.empty:
        story.append(Paragraph("This table could not be read or is empty.", pdf_styles["BodyX"]))
        continue
    tbl, truncated = pdf_table_for_df(df)
    story.append(tbl)
    if truncated:
        story.append(Paragraph(
            "Display limited to first 100 rows and first 8 columns; complete data remain in the CSV/workbook.",
            pdf_styles["BodyX"]
        ))
    story.append(Paragraph("Interpretation", pdf_styles["H2X"]))
    for statement in data_interpretation(path.stem, df):
        story.append(Paragraph(escape(statement), pdf_styles["BodyX"]))
    story.append(Spacer(1, 6))

story.append(PageBreak())
story.append(Paragraph("Figures", pdf_styles["H1X"]))
if not image_files:
    story.append(Paragraph("No figures were found in the configured folders.", pdf_styles["BodyX"]))
for idx, path in enumerate(image_files, start=1):
    story.append(Paragraph(f"Figure {idx}. {escape(nice_name(path.name))}", pdf_styles["H2X"]))
    try:
        with PILImage.open(path) as im:
            iw, ih = im.size
        maxw = A4[0] - 3.5 * cm
        maxh = A4[1] - 5.5 * cm
        scale = min(maxw / iw, maxh / ih, 1.0)
        img = RLImage(str(path), width=iw * scale, height=ih * scale)
        story.append(KeepTogether([
            img,
            Paragraph(
                "Interpretation: use the title and axes to identify the displayed pattern and compare it "
                "with the relevant descriptive or model results. A figure alone does not establish causality.",
                pdf_styles["CaptionX"]
            ),
            Spacer(1, 6)
        ]))
    except Exception as exc:
        story.append(Paragraph(f"Could not insert image: {escape(str(exc))}", pdf_styles["BodyX"]))

story.append(PageBreak())
story.append(Paragraph("Source code appendix", pdf_styles["H1X"]))
story.append(Paragraph(
    "The following appendix reproduces available Python source files. Code is included as text for "
    "reference; run the original .py scripts in the appropriate order in Colab.", pdf_styles["BodyX"]
))

for part_num in range(1, 7):
    source = code_files[part_num]
    story.append(Paragraph(f"Part {part_num} source code", pdf_styles["H2X"]))
    if source is None:
        story.append(Paragraph(
            f"Source file for Part {part_num} was not found. Upload/copy it into /content and rerun this generator.",
            pdf_styles["BodyX"]
        ))
        continue
    story.append(Paragraph(f"File: {escape(source.name)}", pdf_styles["BodyX"]))
    source_text = source.read_text(encoding="utf-8", errors="replace")
    # Escape source text so markup characters in code are rendered literally.
    for line in source_text.splitlines():
        safe = escape(line.expandtabs(4)) or " "
        story.append(Preformatted(safe, pdf_styles["CodeX"]))

story.append(Paragraph("Reproducibility and limitations", pdf_styles["H1X"]))
story.append(Paragraph(
    "Only artifacts saved to disk can be included automatically. Results that were displayed but not "
    "exported cannot be recovered by this script. Automatically generated interpretations are deliberately "
    "cautious and should be checked against the original regression specification and assumptions before submission. "
    "Statistical significance is not equivalent to causality.", pdf_styles["BodyX"]
))

def add_page_number(canvas, doc_obj):
    canvas.saveState()
    canvas.setFont("Helvetica", 7)
    canvas.setFillColor(colors.grey)
    canvas.drawCentredString(A4[0] / 2, 0.8 * cm, f"Page {doc_obj.page}")
    canvas.restoreState()

pdf_doc = SimpleDocTemplate(
    str(PDF_PATH), pagesize=A4,
    rightMargin=1.5 * cm, leftMargin=1.5 * cm,
    topMargin=1.4 * cm, bottomMargin=1.4 * cm,
    title="Comprehensive Applied Econometrics Project Report"
)
pdf_doc.build(story, onFirstPage=add_page_number, onLaterPages=add_page_number)

print("\nReport generation complete.")
print(f"CSV tables included: {len(csv_files)}")
print(f"Figures included: {len(image_files)}")
print(f"Source code files included: {sum(p is not None for p in code_files.values())}/6")
print(f"Word report: {DOCX_PATH}")
print(f"PDF report:  {PDF_PATH}")
print(f"Inventory:   {SUMMARY_PATH}")

# Automatically download the two report files in Google Colab.
try:
    from google.colab import files
    files.download(str(DOCX_PATH))
    files.download(str(PDF_PATH))
except ImportError:
    print("Not running in Google Colab; use the paths printed above to retrieve the files.")
```
